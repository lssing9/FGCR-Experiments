```python
from __future__ import annotations
import gc
import gzip
import hashlib
import json
import platform
import math
import random
import sys
import re
import time
from dataclasses import asdict, dataclass
from itertools import combinations
from pathlib import Path
from statistics import median
from typing import Hashable, Iterable, Mapping, Sequence,Iterator
import matplotlib.pyplot as plt
import numpy as np
import networkx as nx
import pandas as pd

Node = Hashable
Edge = tuple[Node, Node]
```

```python
# 1. 通用输入检查
# 检查 G 是否为简单 DAG。
def _validate_dag(G: nx.DiGraph) -> None:
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。")
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")

#检查 C 是否为 V 的非空真子集。
def _validate_conditioning_set(
    G: nx.DiGraph,
    C: Iterable[Node],
) -> set[Node]:
    C_set = set(C)
    V_set = set(G.nodes)
    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。")
    return C_set

#检查 order 是否恰好包含 V 中全部节点一次。 require_topological=True 时，还要求 order 是 G 的合法拓扑序。返回节点到顺序位置的映射。
def _validate_order(
    G: nx.DiGraph,
    order: Sequence[Node],
    name: str,
    require_topological: bool,
) -> dict[Node, int]:
    order_list = list(order)
    if len(order_list) != G.number_of_nodes():
        raise ValueError(f"{name} 的长度不等于 G 的节点数。")
    if len(set(order_list)) != len(order_list):
        raise ValueError(f"{name} 中存在重复节点。")
    if set(order_list) != set(G.nodes):
        raise ValueError(f"{name} 必须恰好包含 G 的全部节点。")

    position = {node: index for index, node in enumerate(order_list)}
    if require_topological:
        for source, target in G.edges:
            if position[source] >= position[target]:
                raise ValueError(
                    f"{name} 不是 G 的合法拓扑序："
                    f"{source!r}->{target!r} 与顺序冲突。"
                )

def _edge_sort_key(edge: Edge) -> tuple[str, str]:
    return str(edge[0]), str(edge[1])

def _encode_nodes(nodes: Iterable[Node]) -> str:
    return json.dumps(
        [str(node) for node in sorted(nodes, key=str)],
        ensure_ascii=False,
        separators=(",", ":"),
    )

def _encode_order(order: Sequence[Node]) -> str:
    return json.dumps(
        [str(node) for node in order],
        ensure_ascii=False,
        separators=(",", ":"),
    )

def _encode_edges(edges: Iterable[Edge]) -> str:
    ordered = sorted(edges, key=_edge_sort_key)
    return json.dumps(
        [[str(source), str(target)] for source, target in ordered],
        ensure_ascii=False,
        separators=(",", ":"),
    )

def _sha256_text(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

def _hash_nodes(nodes: Iterable[Node]) -> str:
    return _sha256_text(_encode_nodes(nodes))

def _hash_order(order: Sequence[Node]) -> str:
    return _sha256_text(_encode_order(order))

def _hash_edges(edges: Iterable[Edge]) -> str:
    return _sha256_text(_encode_edges(edges))

def _graph_fingerprint(G: nx.DiGraph) -> str:
    payload = {
        "nodes": [str(node) for node in sorted(G.nodes, key=str)],
        "edges": [
            [str(source), str(target)]
            for source, target in sorted(G.edges, key=_edge_sort_key)
        ],
    }
    return _sha256_text(
        json.dumps(
            payload,
            ensure_ascii=False,
            separators=(",", ":"),
        )
    )

def _derive_seed(base_seed: int, *parts: object) -> int:
    """从根种子与实验标识稳定派生 64 位随机种子。"""
    payload = "|".join([str(base_seed), *(str(part) for part in parts)])
    digest = hashlib.blake2b(
        payload.encode("utf-8"),
        digest_size=8,
    ).digest()
    return int.from_bytes(digest, byteorder="big", signed=False)
```

## FCGR 算法

```python
# 算法 1 的输出
@dataclass(frozen=True)
class Algorithm1Orders:
    beta: tuple[Node, ...]
    alpha_C: tuple[Node, ...]
    alpha_R: tuple[Node, ...]
    alpha: tuple[Node, ...]

# FCGR 统一结果对象
@dataclass
class FCGRResult:
    H: nx.DiGraph
    H_R: nx.DiGraph
    G_R: nx.DiGraph
    orders: Algorithm1Orders
    # 最终 H_R 相对于 G_R 新增的边
    added_edges_R: tuple[Edge, ...]
    # 算法 2 运行统计
    n_swaps: int
    n_reversals: int
    n_cover_additions: int

# 随机拓扑序
def random_topological_order(
    G: nx.DiGraph,
    seed: int,
) -> tuple[Node, ...]:
    """随机化 Kahn 算法：生成可复现的合法拓扑序。"""
    _validate_dag(G)
    rng = random.Random(seed)
    indegree = dict(G.in_degree())
    available = sorted(
        [node for node, degree in indegree.items() if degree == 0],
        key=str,
    )
    order: list[Node] = []

    while available:
        selected_index = rng.randrange(len(available))
        selected_node = available.pop(selected_index)
        order.append(selected_node)

        for child in sorted(G.successors(selected_node), key=str):
            indegree[child] -= 1
            if indegree[child] == 0:
                available.append(child)
        available.sort(key=str)

    if len(order) != G.number_of_nodes():
        raise RuntimeError("拓扑排序未覆盖 G 的全部节点。")
    return tuple(order)

# 算法 1：构造 beta 与 alpha
"""
    构造 FCGR 所需顺序。两种调用方式：
    1. 显式给定 beta：
       construct_algorithm1_orders(G=G,C=C,beta=beta, )
    2. 根据 seed 生成 beta：
       construct_algorithm1_orders(G=G,C=C,seed=1,)
    beta 和 seed 不能同时给定。
"""

def construct_algorithm1_orders(
    G: nx.DiGraph,
    C: Iterable[Node],
    beta: Sequence[Node],
) -> Algorithm1Orders:
    """
    FCGR 算法1：
        (G, C, beta) -> alpha_C、alpha_R -> alpha=(alpha_C, alpha_R)。
    """
    _validate_dag(G)
    C_set = _validate_conditioning_set(G, C)
    beta_list = list(beta)
    _validate_order(
        G,
        beta_list,
        "beta",
        require_topological=True,
    )

    alpha_C = tuple(node for node in beta_list if node in C_set)
    alpha_R = tuple(node for node in beta_list if node not in C_set)
    alpha = alpha_C + alpha_R

    return Algorithm1Orders(
        beta=tuple(beta_list),
        alpha_C=alpha_C,
        alpha_R=alpha_R,
        alpha=alpha,
    )


# 覆盖并反转 Y→Z
# 将当前边 Y→Z 补成 covered edge， 然后反转为 Z→Y。返回覆盖过程中新增的边，不包含反转后的边 Z→Y。

def _cover_and_reverse(
    H: nx.DiGraph,
    Y: Node,
    Z: Node,
) -> tuple[Edge, ...]:
    """先把 Y->Z 补成 covered edge，再反转为 Z->Y。"""
    if not H.has_edge(Y, Z):
        raise ValueError(f"当前图中不存在边 {Y!r}->{Z!r}。")

    pa_Y = set(H.predecessors(Y))
    pa_Z = set(H.predecessors(Z))
    added_edges: list[Edge] = []

    for X in sorted(pa_Y - pa_Z, key=str):
        if not H.has_edge(X, Z):
            H.add_edge(X, Z)
            added_edges.append((X, Z))

    for X in sorted(pa_Z - pa_Y - {Y}, key=str):
        if not H.has_edge(X, Y):
            H.add_edge(X, Y)
            added_edges.append((X, Y))

    if set(H.predecessors(Z)) != set(H.predecessors(Y)) | {Y}:
        raise AssertionError(
            f"边 {Y!r}->{Z!r} 在反转前没有成为 covered edge。"
        )

    H.remove_edge(Y, Z)
    H.add_edge(Z, Y)
    return tuple(added_edges)

# 算法 2
# ==========================================================
"""
    根据 FCGR 算法 2 构造完整图 H。返回：
    H:完整转换图。
    n_swaps:顺序交换次数。
    n_reversals:覆盖反转次数。
    n_cover_additions:覆盖反转过程中新增的边数。
"""
def _construct_full_H(G: nx.DiGraph, beta: Sequence[Node], alpha: Sequence[Node]) -> tuple[nx.DiGraph, int, int, int]:
    # FCGR 算法2的顺序调整与覆盖反转主体。
    beta_current = list(beta)
    alpha_list = list(alpha)
    beta_position = _validate_order(G, beta_current, "beta", require_topological=True)
    alpha_position = _validate_order(G, alpha_list, "alpha", require_topological=False)
    H = G.copy()
    n_swaps = 0
    n_reversals = 0
    n_cover_additions = 0
    # 从目标顺序 alpha 的最右端开始处理
    for Y in reversed(alpha_list):
        while beta_position[Y] < len(beta_current) - 1:
            current_position = beta_position[Y]
            Z = beta_current[current_position + 1]
             # Z 在目标顺序中本来就在 Y 的右侧，
            # 此时 Y 已经到达正确位置
            if alpha_position[Z] > alpha_position[Y]:
                break
            if H.has_edge(Y, Z):
                added_edges = _cover_and_reverse(H, Y, Z)
                n_reversals += 1
                n_cover_additions += len(added_edges)
            else:
                added_edges = tuple()
            # 在当前 beta 中交换相邻节点 Y 和 Z
            beta_current[current_position] = Z
            beta_current[current_position + 1] = Y
            beta_position[Z] = current_position
            beta_position[Y] = current_position + 1
            n_swaps += 1

    if beta_current != alpha_list:
        raise AssertionError(f"算法 2 终止后 beta 不等于 alpha：\nbeta={beta_current}\nalpha={alpha_list}")
    for source, target in H.edges:
        #检查 H 的全部边是否与目标顺序 alpha 一致
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError(f"边 {source!r}->{target!r} 与目标顺序 alpha 不一致。")
    if not nx.is_directed_acyclic_graph(H):
        raise AssertionError("算法 2 输出的完整图 H 不是 DAG。")
    H.name = f"{G.name or 'G'}_H"
    return H, n_swaps, n_reversals, n_cover_additions

#构造算法2的最终输出 H_R。这部分属于算法2计时范围，因为算法2的终点是最终条件图 H_R。
def _finish_fcgr_result(
    G: nx.DiGraph,
    orders: Algorithm1Orders,
    H: nx.DiGraph,
    n_swaps: int,
    n_reversals: int,
    n_cover_additions: int,
) -> FCGRResult:
    G_R = G.subgraph(orders.alpha_R).copy()
    G_R.name = f"{G.name or 'G'}_GR"
    H_R = H.subgraph(orders.alpha_R).copy()
    H_R.name = f"{G.name or 'G'}_HR"
    # 最终条件图中真正新增的 R 内部边
    added_edges_R = tuple(sorted(set(H_R.edges()) - set(G_R.edges()), key=_edge_sort_key))
    return FCGRResult(
        H=H,
        H_R=H_R,
        G_R=G_R,
        orders=orders,
        added_edges_R=added_edges_R,
        n_swaps=n_swaps,
        n_reversals=n_reversals,
        n_cover_additions=n_cover_additions,
    )

#统一调用函数并封装结果，不计时的 FCGR 统一调用函数。
"""
    执行完整 FCGR。
    beta:显式给定的原图合法拓扑序。
    seed:beta=None 时，用于生成可复现随机拓扑序。
"""
def run_fcgr(
    G: nx.DiGraph,
    C: Iterable[Node],
    beta: Sequence[Node],
) -> FCGRResult:
    """不计时地执行完整 FCGR，最终输出字段为 result.H_R。"""
    orders = construct_algorithm1_orders(G=G, C=C, beta=beta)
    H, n_swaps, n_reversals, n_cover_additions = _construct_full_H(
        G=G,
        beta=orders.beta,
        alpha=orders.alpha,
    )
    return _finish_fcgr_result(
        G=G,
        orders=orders,
        H=H,
        n_swaps=n_swaps,
        n_reversals=n_reversals,
        n_cover_additions=n_cover_additions,
    )

def _timed_fcgr_once(
    G: nx.DiGraph,
    C: Iterable[Node],
    beta: Sequence[Node],
) -> tuple[FCGRResult, int, int]:
    """单次 FCGR 分段计时，单位为纳秒。"""
    algorithm1_start = time.perf_counter_ns()
    orders = construct_algorithm1_orders(G=G, C=C, beta=beta)
    algorithm1_end = time.perf_counter_ns()

    algorithm2_start = time.perf_counter_ns()
    H, n_swaps, n_reversals, n_cover_additions = _construct_full_H(
        G=G,
        beta=orders.beta,
        alpha=orders.alpha,
    )
    result = _finish_fcgr_result(
        G=G,
        orders=orders,
        H=H,
        n_swaps=n_swaps,
        n_reversals=n_reversals,
        n_cover_additions=n_cover_additions,
    )
    algorithm2_end = time.perf_counter_ns()

    return (
        result,
        algorithm1_end - algorithm1_start,
        algorithm2_end - algorithm2_start,
    )
```

## ROS 算法

```python
#返回 node 的全部祖先，并包含 node 自身。
def ancestors_with_self(
    G: nx.DiGraph,
    node: Node,
) -> set[Node]:
    if node not in G:
        raise ValueError(f"节点 {node!r} 不在图 G 中。")
    return set(nx.ancestors(G, node)) | {node}

# ROS 算法
def minimal_fill_edge_set(
    G: nx.DiGraph,
    alpha: Sequence[Node],
    C: Iterable[Node],
) -> dict[str, object]:
    """
    ROS 算法：寻找条件化模型的最小填充边集。
        E_alpha_R: ROS 找到的新增有向边，每条边保存为 (source, target)，
        G_alpha_R:G_R 加上 E_alpha_R 后得到的条件模型 minimal I-map。
    """
    # 0. 输入验证
    _validate_dag(G)
    C_set = _validate_conditioning_set(G=G, C=C)
    alpha_list = list(alpha)
    alpha_position = _validate_order( G,  alpha_list, "alpha", require_topological=True,)
    G_original = G  
    G_work = G.copy()
    # 1. 构造 G^C：删除 C 中节点的全部出边
    for conditioned_node in C_set:
        outgoing_edges = list(G_work.out_edges(conditioned_node))
        G_work.remove_edges_from(outgoing_edges)
    
    # 2. 计算 An_{G^C}(C)
    nodes_of_interest: set[Node] = set()
    for conditioned_node in C_set:
        nodes_of_interest.update(ancestors_with_self(G_work, conditioned_node))
    
    # 3. 构造祖先闭包上的诱导子图 G_0
    G_0 = G_work.subgraph(nodes_of_interest).copy()
    G_0.name = f"{G.name or 'G'}_ROS_search"
    
    # 4. 按 alpha 的逆序排列搜索节点
    vertices = [node for node in reversed(alpha_list) if node in nodes_of_interest]
    k = len(vertices)

    # 每条新增边保存为真正的节点元组
    E_alpha_R: list[Edge] = []
 
    # 5. 逐层完成父节点集合
    if k >= 3:
        for current_node in vertices[: k - 2]:
            parents = sorted(G_0.predecessors(current_node), key=alpha_position.__getitem__,)
            if len(parents) >= 2:
                for x, y in combinations(parents, 2):
                    # 已经邻接时不重复添加
                    if G_0.has_edge(x, y) or G_0.has_edge(y, x):
                        continue
                    # 按 alpha 从左向右定向
                    if alpha_position[x] < alpha_position[y]: 
                         new_edge = (x, y)
                    else:
                         new_edge = (y, x)
                    G_0.add_edge(*new_edge)
                    E_alpha_R.append(new_edge)
            G_0.remove_node(current_node)

    # 6. 构造原图在 R=V\\C 上的完整导出子图
    R_nodes = [node for node in G_original.nodes if node not in C_set]
    G_alpha_R = G_original.subgraph(R_nodes).copy()
    G_alpha_R.name = f"{G.name or 'G'}_ROS_HR"
    
    # 7. 添加 ROS 找到的填充边
    G_alpha_R.add_edges_from(E_alpha_R)
    
    # 8. 输出检查
    if not nx.is_directed_acyclic_graph(G_alpha_R):
        raise AssertionError("ROS 输出图 G_alpha_R 不是 DAG。")

    for source, target in G_alpha_R.edges:
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError("ROS 输出边与 alpha 不一致：" f"{source!r}->{target!r}。")

    return { "E_alpha_R": tuple( sorted(set(E_alpha_R), key=_edge_sort_key)),"G_alpha_R": G_alpha_R,}

# 单次 ROS 总计时。
def _timed_ros_once(
    G: nx.DiGraph,
    alpha: Sequence[Node],
    C: Iterable[Node],
) -> tuple[dict[str, object], int]:
    start = time.perf_counter_ns()
    result = minimal_fill_edge_set(G=G, alpha=alpha, C=C)
    end = time.perf_counter_ns()
    return result, end - start
```

## 标准真实网络实验定义

```python
# 标准网络结构读取
EXPECTED_NODE_COUNTS = {
    "Asia": 8,
    "Insurance": 27,
    "Alarm": 37,
    "Win95pts": 76,
    "Andes": 223,
    "Munin": 1041,
}

def _clean_bif_name(raw_name: str) -> str:
    """清理 BIF 文件中的变量名。"""
    name = raw_name.strip()
    if len(name) >= 2 and name[0] == name[-1] and name[0] in {"'", '"'}:
        name = name[1:-1]
    return name.strip()

def _load_bif_structure_fallback(path: Path) -> nx.DiGraph:
    """
    无 pgmpy 时，只读取 BIF 的变量名和父子关系。
    """
    text = path.read_text(encoding="utf-8", errors="ignore")
    text = re.sub(r"/\*.*?\*/", "", text, flags=re.S)
    text = re.sub(r"//.*?$", "", text, flags=re.M)
    G = nx.DiGraph(name=path.stem)
    variable_pattern = re.compile(r"\bvariable\s+(\"[^\"]+\"|'[^']+'|[^\s\{]+)\s*\{", flags=re.I)
    for match in variable_pattern.finditer(text):
        node = _clean_bif_name(match.group(1))
        if node:
            G.add_node(node)
    probability_pattern = re.compile(r"\bprobability\s*\(\s*(.*?)\s*\)\s*\{", flags=re.I | re.S)
    for match in probability_pattern.finditer(text):
        header = match.group(1).strip()
        if "|" in header:
            child_text, parent_text = header.split("|", 1)
            parents = [_clean_bif_name(item) for item in parent_text.split(",") if _clean_bif_name(item)]
        else:
            child_text = header
            parents = []
        child = _clean_bif_name(child_text)
        if not child:
            continue
        G.add_node(child)
        for parent in parents:
            G.add_edge(parent, child)
    if G.number_of_nodes() == 0:
        raise ValueError(f"未能从 BIF 文件读取任何节点：{path}")
    return G

def load_bif_dag(path: str | Path) -> nx.DiGraph:
    """从 BIF 文件读取 DAG。"""
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"BIF 文件不存在：{path}")
    try:
        from pgmpy.readwrite import BIFReader
    except ImportError:
        G = _load_bif_structure_fallback(path)
    else:
        model = BIFReader(str(path)).get_model()
        G = nx.DiGraph(name=path.stem)
        G.add_nodes_from(model.nodes())
        G.add_edges_from(model.edges())
    _validate_dag(G)
    return G

def load_standard_networks(data_dir: str | Path, network_files: Mapping[str, str], strict_node_count: bool = True) -> dict[str, nx.DiGraph]:
    """读取六个标准贝叶斯网络。"""
    data_dir = Path(data_dir)
    graphs: dict[str, nx.DiGraph] = {}
    for network, filename in network_files.items():
        path = data_dir / filename
        G = load_bif_dag(path)
        G.name = network
        expected_n = EXPECTED_NODE_COUNTS.get(network)
        if strict_node_count and expected_n is not None and G.number_of_nodes() != expected_n:
            raise ValueError(f"{network} 节点数不正确：读取到 {G.number_of_nodes()}，预期 {expected_n}。")
        graphs[network] = G
    return graphs
```

## 标准网络实验框架

```python
# 实验 A：720 个边集与输出图一致性实例
#每个网络使用30个三节点条件集和4个固定拓扑序。
# 实例总数：6个网络 * 30个条件集 * 4个拓扑序 = 720。
@dataclass(frozen=True)
class StandardConsistencyConfig:
    c_size: int = 3
    condition_sets_per_network: int = 30
    n_fixed_orders: int = 4
    # 控制条件集，固定后条件集不变
    condition_base_seed: int = 20260711
    # 控制4个随机拓扑序
    order_base_seed: int = 20260811
    stop_on_mismatch: bool = False

def _validate_consistency_config(config: StandardConsistencyConfig) -> None:
    if config.c_size <= 0:
        raise ValueError("c_size 必须为正整数。")
    if config.condition_sets_per_network <= 0:
        raise ValueError("condition_sets_per_network 必须为正整数。")
    if config.n_fixed_orders <= 0:
        raise ValueError("n_fixed_orders 必须为正整数。")

#生成可复现且互不相同的固定合法拓扑序。
def generate_fixed_random_topological_orders(G: nx.DiGraph, n_orders: int, seed_base: int, max_attempts: int = 10_000) -> list[dict[str, object]]:
    if n_orders <= 0:
        raise ValueError("n_orders 必须为正整数。")
    orders: list[dict[str, object]] = []
    seen: set[tuple[Node, ...]] = set()
    attempt = 0
    while len(orders) < n_orders and attempt < max_attempts:
        beta_seed = seed_base + attempt
        beta = tuple(random_topological_order(G=G, seed=beta_seed))
        if beta not in seen:
            seen.add(beta)
            orders.append({"order_id": len(orders) + 1, "beta_seed": beta_seed, "beta": beta})
        attempt += 1
    if len(orders) < n_orders:
        raise RuntimeError(f"只能生成 {len(orders)} 个不同拓扑序，未达到要求的 {n_orders} 个。")
    return orders

#在同一个 (G,C,beta) 下配对运行FCGR与ROS。
#验证： 1. 有向填充边集是否一致； 2. 最终输出图的节点集和完整有向边集是否一致。
def run_one_consistency_case(network: str, G: nx.DiGraph, condition_id: int, order_id: int, C: Sequence[Node], beta: Sequence[Node], condition_seed: int, beta_seed: int) -> dict[str, object]:
    C_set = set(C)
    row: dict[str, object] = {
        "network": network,
        "condition_id": condition_id,
        "order_id": order_id,
        "condition_seed": condition_seed,
        "beta_seed": beta_seed,
        "n_nodes": G.number_of_nodes(),
        "n_edges": G.number_of_edges(),
        "c_size": len(C_set),
        "C": _encode_nodes(C_set),
        "beta": _encode_order(beta),
        "status": "ok",
        "error_type": "",
        "error_message": "",
    }
    try:
        fcgr_result = run_fcgr(G=G.copy(), C=C_set, beta=beta)
        ros_result = minimal_fill_edge_set(G=G.copy(), alpha=beta, C=C_set)
        fcgr_fill = set(fcgr_result.added_edges_R)
        ros_fill = set(ros_result["E_alpha_R"])
        ros_graph = ros_result["G_alpha_R"]
        if not isinstance(ros_graph, nx.DiGraph):
            raise TypeError("ROS结果中的G_alpha_R必须是nx.DiGraph。")
        fcgr_nodes = set(fcgr_result.H_R.nodes())
        ros_nodes = set(ros_graph.nodes())
        fcgr_graph_edges = set(fcgr_result.H_R.edges())
        ros_graph_edges = set(ros_graph.edges())
        fill_equal = fcgr_fill == ros_fill
        graph_equal = fcgr_nodes == ros_nodes and fcgr_graph_edges == ros_graph_edges
        row.update({
            "FCGR_fill_count": len(fcgr_fill),
            "ROS_fill_count": len(ros_fill),
            "FCGR_output_edge_count": len(fcgr_graph_edges),
            "ROS_output_edge_count": len(ros_graph_edges),
            "fill_equal": fill_equal,
            "graph_equal": graph_equal,
            "both_equal": fill_equal and graph_equal,
            "FCGR_fill_edges": _encode_edges(fcgr_fill),
            "ROS_fill_edges": _encode_edges(ros_fill),
            "only_FCGR_fill_edges": _encode_edges(fcgr_fill - ros_fill),
            "only_ROS_fill_edges": _encode_edges(ros_fill - fcgr_fill),
            "only_FCGR_output_edges": _encode_edges(fcgr_graph_edges - ros_graph_edges),
            "only_ROS_output_edges": _encode_edges(ros_graph_edges - fcgr_graph_edges),
        })
    except Exception as exc:
        row.update({
            "status": "error",
            "error_type": type(exc).__name__,
            "error_message": str(exc),
            "fill_equal": False,
            "graph_equal": False,
            "both_equal": False,
        })
    return row

def _summarize_consistency(details: pd.DataFrame) -> pd.DataFrame:
    summary_rows: list[dict[str, object]] = []
    for network, group in details.groupby("network", sort=False):
        successful = group[group["status"] == "ok"]
        n_total = len(group)
        n_successful = len(successful)
        def rate_percent(column: str) -> float:
            values = (group[column].fillna(False) .astype(bool) )
            return 100.0 * float(values.mean())
        mismatch_mask = (group["status"] != "ok") | (~group["fill_equal"].fillna(False)) | (~group["graph_equal"].fillna(False))
        summary_rows.append({
            "network": network,
            "n_nodes": int(group["n_nodes"].iloc[0]),
            "n_edges": int(group["n_edges"].iloc[0]),
            "c_size": int(group["c_size"].iloc[0]),
            "n_condition_sets": group["condition_id"].nunique(),
            "n_fixed_orders": group["order_id"].nunique(),
            "n_instances": n_total,
            "n_successful": n_successful,
            "n_errors": n_total - n_successful,
            "fill_equal_rate_pct": rate_percent("fill_equal"),
            "graph_equal_rate_pct": rate_percent("graph_equal"),
            "both_equal_rate_pct": rate_percent("both_equal"),
            "n_mismatches": int(mismatch_mask.sum()),
        })
    return pd.DataFrame(summary_rows)

# 完整运行720个一致性实例。
#每个网络的30个条件集与4个固定拓扑序完全交叉。
def run_standard_consistency_experiment(network_graphs: Mapping[str, nx.DiGraph], config: StandardConsistencyConfig = StandardConsistencyConfig()) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    _validate_consistency_config(config)
    rows: list[dict[str, object]] = []
    order_rows: list[dict[str, object]] = []
    for network_index, (network, G) in enumerate(network_graphs.items(), start=1):
        _validate_dag(G)
        if not 1 <= config.c_size < G.number_of_nodes():
            raise ValueError(f"{network}不满足1 <= c_size < |V|。")
        condition_seed = config.condition_base_seed + network_index * 1_000_000 + 101
        order_seed_base = config.order_base_seed + network_index * 1_000_000 + 10_000
        condition_sets = sample_random_condition_sets(G=G, n_sets=config.condition_sets_per_network, c_size=config.c_size, seed=condition_seed)
        if len(set(condition_sets)) != config.condition_sets_per_network:
            raise AssertionError(f"{network}未生成30个互不相同的条件集。")
        fixed_orders = generate_fixed_random_topological_orders(G=G, n_orders=config.n_fixed_orders, seed_base=order_seed_base)
        for item in fixed_orders:
            order_rows.append({"network": network, "order_id": item["order_id"], "beta_seed": item["beta_seed"], "beta": _encode_order(item["beta"])})
        for condition_id, C in enumerate(condition_sets, start=1):
            for order_info in fixed_orders:
                row = run_one_consistency_case(
                    network=network, G=G, condition_id=condition_id, order_id=int(order_info["order_id"]),
                    C=C, beta=order_info["beta"], condition_seed=condition_seed, beta_seed=int(order_info["beta_seed"])
                )
                rows.append(row)
                if config.stop_on_mismatch and not row["both_equal"]:
                    raise AssertionError(f"{network}，条件集{condition_id}，拓扑序{order_info['order_id']}未通过一致性检查。")
        print(f"[一致性][{network}] 完成{len(condition_sets) * len(fixed_orders)}个实例。")
    details = pd.DataFrame(rows)
    fixed_orders_df = pd.DataFrame(order_rows)
    mismatch_mask = (details["status"] != "ok") | (~details["fill_equal"].fillna(False)) | (~details["graph_equal"].fillna(False))
    mismatches = details.loc[mismatch_mask].copy()
    summary = _summarize_consistency(details)
    expected_instances = len(network_graphs) * config.condition_sets_per_network * config.n_fixed_orders
    if len(details) != expected_instances:
        raise AssertionError(f"一致性实验实例数错误：得到{len(details)}，预期{expected_instances}。")
    return details, summary, mismatches, fixed_orders_df

#保存720实例一致性实验的全部结果。
def save_consistency_results(details: pd.DataFrame, summary: pd.DataFrame, mismatches: pd.DataFrame, fixed_orders: pd.DataFrame, config: StandardConsistencyConfig, output_file: str | Path) -> Path:
    output_file = Path(output_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)
    config_df = pd.DataFrame({"parameter": list(asdict(config).keys()), "value": list(asdict(config).values())})
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(writer, sheet_name="network_summary", index=False)
        fixed_orders.to_excel(writer, sheet_name="fixed_orders", index=False)
        details.to_excel(writer, sheet_name="instance_results", index=False)
        mismatches.to_excel(writer, sheet_name="mismatches", index=False)
        config_df.to_excel(writer, sheet_name="config", index=False)
    return output_file
```

```python
# 实验 B：180 个独立运行时间实例
@dataclass(frozen=True)
class RuntimeExperimentConfig:
    c_size: int = 3
    condition_sets_per_network: int = 30
    warmup_repetitions: int = 5
    timing_repetitions: int = 30
    condition_base_seed: int = 20260711
    order_base_seed: int = 20260811
    stop_on_mismatch: bool = True

def _validate_runtime_config(config: RuntimeExperimentConfig) -> None:
    if config.c_size <= 0:
        raise ValueError("c_size 必须为正整数。")
    if config.condition_sets_per_network <= 0:
        raise ValueError("condition_sets_per_network 必须为正整数。")
    if config.warmup_repetitions < 0:
        raise ValueError("warmup_repetitions 不能为负数。")
    if config.timing_repetitions <= 0:
        raise ValueError("timing_repetitions 必须为正整数。")

def _outputs_are_equal(fcgr_result: FCGRResult, ros_result: Mapping[str, object]) -> tuple[bool, bool]:
    fcgr_fill = set(fcgr_result.added_edges_R)
    ros_fill = set(ros_result["E_alpha_R"])
    ros_graph = ros_result["G_alpha_R"]
    if not isinstance(ros_graph, nx.DiGraph):
        raise TypeError("ROS 结果中的 G_alpha_R 必须是 nx.DiGraph。")
    fill_equal = fcgr_fill == ros_fill
    graph_equal = set(fcgr_result.H_R.nodes()) == set(ros_graph.nodes()) and set(fcgr_result.H_R.edges()) == set(ros_graph.edges())
    return fill_equal, graph_equal

def _run_pair_without_gc(G: nx.DiGraph, C: Sequence[Node], beta: Sequence[Node], fcgr_first: bool) -> tuple[FCGRResult, dict[str, object], int, int, int]:
    """
    在自动垃圾回收关闭时运行一对计时。
    返回：
        FCGR结果、ROS结果、算法1 ns、算法2 ns、ROS总 ns。
    """
    gc_was_enabled = gc.isenabled()
    if gc_was_enabled:
        gc.disable()
    try:
        if fcgr_first:
            fcgr_result, algorithm1_ns, algorithm2_ns = _timed_fcgr_once(G=G, C=C, beta=beta)
            ros_result, ros_total_ns = _timed_ros_once(G=G, alpha=beta, C=C)
        else:
            ros_result, ros_total_ns = _timed_ros_once(G=G, alpha=beta, C=C)
            fcgr_result, algorithm1_ns, algorithm2_ns = _timed_fcgr_once(G=G, C=C, beta=beta)
    finally:
        if gc_was_enabled:
            gc.enable()
    return fcgr_result, ros_result, algorithm1_ns, algorithm2_ns, ros_total_ns

def _warm_up_network(G: nx.DiGraph, C: Sequence[Node], beta: Sequence[Node], repetitions: int) -> None:
    """预热不记录时间，但检查输出一致性。"""
    for warmup_id in range(1, repetitions + 1):
        gc.collect()
        fcgr_result, ros_result, _, _, _ = _run_pair_without_gc(G=G, C=C, beta=beta, fcgr_first=warmup_id % 2 == 1)
        fill_equal, graph_equal = _outputs_are_equal(fcgr_result, ros_result)
        if not (fill_equal and graph_equal):
            raise AssertionError(f"{G.name} 预热阶段 FCGR 与 ROS 输出不一致。")

def run_one_runtime_case(network: str, G: nx.DiGraph, condition_id: int, C: Sequence[Node], beta: Sequence[Node], condition_seed: int, beta_seed: int, timing_repetitions: int, stop_on_mismatch: bool) -> tuple[dict[str, object], list[dict[str, object]]]:
    """
    对固定 (G,C,beta) 重复计时，并以内部重复的中位数代表该条件集。
    """
    algorithm1_times_ns: list[int] = []
    algorithm2_times_ns: list[int] = []
    fcgr_total_times_ns: list[int] = []
    ros_total_times_ns: list[int] = []
    repeat_rows: list[dict[str, object]] = []
    all_fill_equal = True
    all_graph_equal = True
    last_fcgr_result: FCGRResult | None = None
    last_ros_result: dict[str, object] | None = None
    for repeat_id in range(1, timing_repetitions + 1):
        gc.collect()
        fcgr_first = (condition_id + repeat_id) % 2 == 0
        fcgr_result, ros_result, algorithm1_ns, algorithm2_ns, ros_total_ns = _run_pair_without_gc(G=G, C=C, beta=beta, fcgr_first=fcgr_first)
        fill_equal, graph_equal = _outputs_are_equal(fcgr_result, ros_result)
        all_fill_equal = all_fill_equal and fill_equal
        all_graph_equal = all_graph_equal and graph_equal
        if stop_on_mismatch and not (fill_equal and graph_equal):
            raise AssertionError(f"{network}，条件集 {condition_id}，第 {repeat_id} 次计时的输出不一致。")
        algorithm1_times_ns.append(algorithm1_ns)
        algorithm2_times_ns.append(algorithm2_ns)
        fcgr_total_ns = (algorithm1_ns+ algorithm2_ns)
        fcgr_total_times_ns.append( fcgr_total_ns)
        ros_total_times_ns.append(ros_total_ns)
        repeat_rows.append({
            "network": network,
            "condition_id": condition_id,
            "repeat_id": repeat_id,
            "run_order": "FCGR_then_ROS" if fcgr_first else "ROS_then_FCGR",
            "FCGR_algorithm1_ms": algorithm1_ns / 1_000_000,
            "FCGR_algorithm2_ms": algorithm2_ns / 1_000_000,
            "FCGR_total_ms": (algorithm1_ns + algorithm2_ns) / 1_000_000,
            "ROS_total_ms": ros_total_ns / 1_000_000,
            "fill_equal": fill_equal,
            "graph_equal": graph_equal,
        })
        last_fcgr_result = fcgr_result
        last_ros_result = ros_result
    if last_fcgr_result is None or last_ros_result is None:
        raise RuntimeError("该条件集没有产生任何正式计时结果。")
    algorithm1_median_ns = float(median(algorithm1_times_ns))
    algorithm2_median_ns = float(median(algorithm2_times_ns))
    fcgr_total_median_ns = float(median(fcgr_total_times_ns))
    ros_total_median_ns = float(median(ros_total_times_ns))
    fcgr_fill_count = len(last_fcgr_result.added_edges_R)
    ros_fill_count = len(last_ros_result["E_alpha_R"])
    condition_row: dict[str, object] = {
        "network": network,
        "condition_id": condition_id,
        "condition_seed": condition_seed,
        "beta_seed": beta_seed,
        "n_nodes": G.number_of_nodes(),
        "n_edges": G.number_of_edges(),
        "c_size": len(C),
        "C": _encode_nodes(C),
        "timing_repetitions": timing_repetitions,
        "FCGR_algorithm1_median_ms": algorithm1_median_ns / 1_000_000,
        "FCGR_algorithm2_median_ms": algorithm2_median_ns / 1_000_000,
        "FCGR_total_median_ms": fcgr_total_median_ns / 1_000_000,
        "ROS_total_median_ms": ros_total_median_ns / 1_000_000,
        "Ratio of Runtime": fcgr_total_median_ns / ros_total_median_ns,
        "FCGR_fill_count": fcgr_fill_count,
        "ROS_fill_count": ros_fill_count,
        "fill_equal": all_fill_equal,
        "graph_equal": all_graph_equal,
        "both_equal": all_fill_equal and all_graph_equal,
    }
    return condition_row, repeat_rows

def _summarize_runtime(details: pd.DataFrame) -> pd.DataFrame:
    summary_rows: list[dict[str, object]] = []
    for network, group in details.groupby("network", sort=False):
        fcgr_algorithm1_mean = float(group["FCGR_algorithm1_median_ms"].mean())
        fcgr_algorithm2_mean = float(group["FCGR_algorithm2_median_ms"].mean())
        fcgr_total_mean = float( group["FCGR_total_median_ms"].mean())
        ros_total_mean = float(group["ROS_total_median_ms"].mean())
        summary_rows.append({
            "network": network,
            "n_nodes": int(group["n_nodes"].iloc[0]),
            "n_edges": int(group["n_edges"].iloc[0]),
            "c_size": int(group["c_size"].iloc[0]),
            "n_condition_sets": len(group),
            "n_unique_condition_sets": group["C"].nunique(),
            "FCGR_algorithm1_mean_ms": fcgr_algorithm1_mean,
            "FCGR_algorithm2_mean_ms": fcgr_algorithm2_mean,
            "FCGR_total_mean_ms": fcgr_total_mean,
            "ROS_total_mean_ms": ros_total_mean,
            "Ratio of Runtime": fcgr_total_mean / ros_total_mean,
            "all_fill_equal": bool(group["fill_equal"].all()),
            "all_graph_equal": bool(group["graph_equal"].all()),
            "all_outputs_equal": bool(group["both_equal"].all()),
        })
    return pd.DataFrame(summary_rows)

def run_runtime_experiment(network_graphs: Mapping[str, nx.DiGraph], config: RuntimeExperimentConfig = RuntimeExperimentConfig()) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    运行六网络配对时间实验。
    返回：
        network_summary：每个网络对30个条件查询求平均后的主结果；
        condition_details：每个条件集的内部重复中位时间；
        repeat_times：每次原始计时；
        fixed_inputs：每个网络固定拓扑序、随机种子和条件集数量。
    """
    _validate_runtime_config(config)
    condition_rows: list[dict[str, object]] = []
    repeat_rows: list[dict[str, object]] = []
    fixed_input_rows: list[dict[str, object]] = []
    for network_index, (network, G) in enumerate(network_graphs.items(), start=1):
        _validate_dag(G)
        if not 1 <= config.c_size < G.number_of_nodes():
            raise ValueError(f"{network} 不满足 1 <= c_size < |V|。")
        condition_seed = config.condition_base_seed + network_index * 1_000_000 + 101
        beta_seed = config.order_base_seed + network_index * 1_000_000 + 10_000
        beta = tuple(random_topological_order(G, seed=beta_seed))
        condition_sets = sample_random_condition_sets(G=G, n_sets=config.condition_sets_per_network, c_size=config.c_size, seed=condition_seed)
        if len(set(condition_sets)) != config.condition_sets_per_network:
            raise AssertionError(f"{network}未生成30个互不相同的条件集。")
        fixed_input_rows.append({
            "network": network,
            "n_nodes": G.number_of_nodes(),
            "n_edges": G.number_of_edges(),
            "condition_seed": condition_seed,
            "beta_seed": beta_seed,
            "c_size": config.c_size,
            "n_condition_sets": len(condition_sets),
            "n_unique_condition_sets": len(set(condition_sets)),
            "beta": _encode_order(beta),
        })
        _warm_up_network(G=G, C=condition_sets[0], beta=beta, repetitions=config.warmup_repetitions)
        for condition_id, C in enumerate(condition_sets, start=1):
            condition_row, current_repeat_rows = run_one_runtime_case(
                network=network, G=G, condition_id=condition_id, C=C, beta=beta,
                condition_seed=condition_seed, beta_seed=beta_seed,
                timing_repetitions=config.timing_repetitions, stop_on_mismatch=config.stop_on_mismatch
            )
            condition_rows.append(condition_row)
            repeat_rows.extend(current_repeat_rows)
        current_details = pd.DataFrame(row for row in condition_rows if row["network"] == network)
        current_summary = _summarize_runtime(current_details).iloc[0]
        print(f"[计时][{network}] 完成{len(condition_sets)}个条件集实例；FCGR={current_summary['FCGR_total_mean_ms']:.6f} ms，ROS={current_summary['ROS_total_mean_ms']:.6f} ms，输出一致={current_summary['all_outputs_equal']}。")
    details = pd.DataFrame(condition_rows)
    repeat_times = pd.DataFrame(repeat_rows)
    fixed_inputs = pd.DataFrame(fixed_input_rows)
    summary = _summarize_runtime(details)
    expected_instances = len(network_graphs) * config.condition_sets_per_network
    if len(details) != expected_instances:
        raise AssertionError(f"计时实验实例数错误：得到{len(details)}，预期{expected_instances}。")
    return summary, details, repeat_times, fixed_inputs

def save_runtime_results(summary: pd.DataFrame, details: pd.DataFrame, repeat_times: pd.DataFrame, fixed_inputs: pd.DataFrame, config: RuntimeExperimentConfig, output_file: str | Path) -> Path:
    """将主表、条件集级结果、原始计时、固定输入和配置保存到 Excel。"""
    output_file = Path(output_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)
    config_df = pd.DataFrame({"parameter": list(asdict(config).keys()), "value": list(asdict(config).values())})
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(writer, sheet_name="network_summary", index=False)
        details.to_excel(writer, sheet_name="condition_details", index=False)
        repeat_times.to_excel(writer, sheet_name="raw_repeat_times", index=False)
        fixed_inputs.to_excel(writer, sheet_name="fixed_inputs", index=False)
        config_df.to_excel(writer, sheet_name="config", index=False)
    return output_file
```

```python
#  六个标准网络的正式配置与运行入口
DATA_DIR = Path(r"D:\jupyternoteboook\条件化实验\ROS and\data")

NETWORK_FILES = {"Asia": "asia.bif", "Insurance": "insurance.bif", "Alarm": "alarm.bif", "Win95pts": "win95pts.bif", "Andes": "andes.bif", "Munin": "munin.bif"}

CONSISTENCY_CONFIG = StandardConsistencyConfig(c_size=3, condition_sets_per_network=30, n_fixed_orders=4, condition_base_seed=20260729, order_base_seed=20260812, stop_on_mismatch=False)

RUNTIME_CONFIG = RuntimeExperimentConfig(c_size=3, condition_sets_per_network=30, warmup_repetitions=2, timing_repetitions=10, condition_base_seed=20260729, order_base_seed=20260812, stop_on_mismatch=True)

CONSISTENCY_RESULT_FILE = Path("./results/FCGR_ROS_consistency.xlsx")
RUNTIME_RESULT_FILE = Path("./results/FCGR_ROS_runtime.xlsx")

def main() -> None:
    network_graphs = load_standard_networks(data_dir=DATA_DIR, network_files=NETWORK_FILES, strict_node_count=True)
    consistency_details, consistency_summary, consistency_mismatches, fixed_orders = run_standard_consistency_experiment(network_graphs=network_graphs, config=CONSISTENCY_CONFIG)
    consistency_saved_path = save_consistency_results(details=consistency_details, summary=consistency_summary, mismatches=consistency_mismatches, fixed_orders=fixed_orders, config=CONSISTENCY_CONFIG, output_file=CONSISTENCY_RESULT_FILE)
    summary, details, repeat_times, fixed_inputs = run_runtime_experiment(network_graphs=network_graphs, config=RUNTIME_CONFIG)
    runtime_saved_path = save_runtime_results(summary=summary, details=details, repeat_times=repeat_times, fixed_inputs=fixed_inputs, config=RUNTIME_CONFIG, output_file=RUNTIME_RESULT_FILE)
    print("\n720个一致性实例的网络汇总：")
    print(consistency_summary.to_string(index=False))
    print(f"不一致或报错实例数：{len(consistency_mismatches)}")
    print("\n180个计时实例的网络平均运行时间：")
    print(summary.to_string(index=False))
    print(f"\n一致性结果已保存到：{consistency_saved_path.resolve()}")
    print(f"计时结果已保存到：{runtime_saved_path.resolve()}")

if __name__ == "__main__":
    main()
```
