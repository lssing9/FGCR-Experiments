# FCGR 与 ROS：六个标准贝叶斯网络的结构一致性实验

实验网络：Asia、Insurance、Alarm、Win95pts、Andes、Munin。

每个配对实例执行：

1. 从网络节点中均匀随机抽取二元条件集 \(C\)；
2. 随机生成原始 DAG \(G\) 的合法基础拓扑序 \(eta\)；
3. FCGR 与 ROS 使用完全相同的 \((G,C,eta)\)；
4. 比较填充边集、最终输出图、逐节点父集和基本合法性。

本 Notebook 暂不测量运行时间，只验证结构输出是否一致。


```python
import json
import math
import random
import re
from dataclasses import dataclass
from pathlib import Path
from typing import Hashable, Iterable, Mapping, Sequence

import networkx as nx
import pandas as pd
from IPython.display import display

Node = Hashable
Edge = tuple[Node, Node]
```

## FCGR 算法

```python
# 算法 1 的输出
@dataclass(frozen=True)
class Algorithm1Orders:
    """
    FCGR 算法 1 产生的全部顺序。
    """
    beta: tuple[Node, ...]
    alpha_C: tuple[Node, ...]
    alpha_R: tuple[Node, ...]
    alpha: tuple[Node, ...]
    seed: int | None = None

# FCGR 统一结果对象
@dataclass
class FCGRResult:
    H: nx.DiGraph
    H_R: nx.DiGraph
    G_R: nx.DiGraph
    orders: Algorithm1Orders
    # 最终 H_R 相对于 G_R 新增的边
    added_edges_R: tuple[Edge, ...]
    # 算法 2 运行统计
    n_swaps: int
    n_reversals: int
    n_cover_additions: int


# 通用输入检查


def _validate_dag(G: nx.DiGraph) -> None:
    """
    检查 G 是否为简单 DAG。
    """
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。"  )
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")


def _validate_conditioning_set(G: nx.DiGraph, C: Iterable[Node]) -> set[Node]:
    """
    检查 C 是否为 V 的非空真子集。
    """
    C_set = set(C)
    V_set = set(G.nodes)
    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。" )
    return C_set


def _validate_order(G: nx.DiGraph, order: Sequence[Node], name: str, require_topological: bool = False) -> dict[Node, int]:
    """
    检查 order 是否恰好包含 V 中全部节点一次。
    require_topological=True 时，
    还要求 order 是 G 的合法拓扑序。
    返回节点到顺序位置的映射。
    """
    order_list = list(order)
    V_set = set(G.nodes)
    if len(order_list) != G.number_of_nodes() or set(order_list) != V_set:
        raise ValueError(f"{name} 必须恰好包含 V 中的每个节点一次。")
    position = {node: index for index, node in enumerate(order_list)}
    if require_topological:
        for source, target in G.edges:
            if position[source] >= position[target]:
                raise ValueError(  f"{name} 不是 G 的合法拓扑序：" f"边 {source!r}->{target!r} " "与该顺序冲突。" )
    return position


def _edge_sort_key(
    edge: Edge,
) -> tuple[str, str]:
    """
    支持字符串、整数等异构节点的稳定边排序。
    """
    return str(edge[0]), str(edge[1])        

# ==========================================================
# 随机拓扑序
# ==========================================================
def random_topological_order(G: nx.DiGraph, seed: int | None = None) -> list[Node]:
    """
    生成 G 的一个合法拓扑序。
    seed=None：
        每次选择按字符串排序后的第一个零入度节点，
        因此结果确定。
    seed为整数：
        在当前零入度节点中随机选择，且结果可复现。
    """
    _validate_dag(G)
    rng = random.Random(seed)
    indegree = dict(G.in_degree())
    available = sorted([node for node, degree in indegree.items() if degree == 0], key=str)
    beta: list[Node] = []
    while available:
        selected_index = 0 if seed is None else rng.randrange(len(available))
        selected_node = available.pop(selected_index)
        beta.append(selected_node)
        for child in sorted(G.successors(selected_node), key=str):
            indegree[child] -= 1
            if indegree[child] == 0:
                available.append(child)
        available.sort(key=str)
    if len(beta) != G.number_of_nodes():
        raise RuntimeError("拓扑排序未覆盖 G 的全部节点。")
    return beta


# ==========================================================
# 算法 1：构造 beta 与 alpha
"""
    构造 FCGR 所需顺序。
    两种调用方式
    ----------
    1. 显式给定 beta：
       construct_algorithm1_orders(
           G=G,
           C=C,
           beta=beta,
       )
    2. 根据 seed 生成 beta：
       construct_algorithm1_orders(
           G=G,
           C=C,
           seed=1,
       )
    beta 和 seed 不能同时给定。
"""
def construct_algorithm1_orders(G: nx.DiGraph, C: Iterable[Node], beta: Sequence[Node] | None = None, seed: int | None = None) -> Algorithm1Orders:
    _validate_dag(G)
    C_set = _validate_conditioning_set(G, C)
    if beta is not None and seed is not None:
        raise ValueError("beta 和 seed 不能同时指定。")
    if beta is None:
        beta_list = random_topological_order(G, seed=seed)
        result_seed = seed
    else:
        beta_list = list(beta)
        _validate_order(G, beta_list, "beta", require_topological=True)
        result_seed = None

    alpha_C = [node for node in beta_list if node in C_set]
    alpha_R = [node for node in beta_list if node not in C_set]
    alpha = alpha_C + alpha_R

    return Algorithm1Orders(
        beta=tuple(beta_list),
        alpha_C=tuple(alpha_C),
        alpha_R=tuple(alpha_R),
        alpha=tuple(alpha),
        seed=result_seed,
    )


# 覆盖并反转 Y→Z
"""
    将当前边 Y→Z 补成 covered edge，
    然后反转为 Z→Y。
    返回覆盖过程中新增的边，
    不包含反转后的边 Z→Y。
"""
# ==========================================================
def _cover_and_reverse(H: nx.DiGraph, Y: Node, Z: Node) -> tuple[Edge, ...]:
    if not H.has_edge(Y, Z):
        raise ValueError(f"当前图中不存在边 {Y!r}->{Z!r}。")
    pa_Y = set(H.predecessors(Y))
    pa_Z = set(H.predecessors(Z))
    added_edges: list[Edge] = []
    # 将 pa(Y)\\pa(Z) 补成 Z 的父节点
    for X in sorted(pa_Y - pa_Z, key=str):
        if not H.has_edge(X, Z):
            H.add_edge(X, Z)
            added_edges.append((X, Z))
    # 将 pa(Z)\\(pa(Y)∪{Y}) 补成 Y 的父节点
    for X in sorted(pa_Z - pa_Y - {Y}, key=str):
        if not H.has_edge(X, Y):
            H.add_edge(X, Y)
            added_edges.append((X, Y))
    # 覆盖边条件：
    # pa(Z) = pa(Y) ∪ {Y}
    if set(H.predecessors(Z)) != set(H.predecessors(Y)) | {Y}:
        raise AssertionError(f"边 {Y!r}->{Z!r} 在反转前未成为 covered edge。")

    H.remove_edge(Y, Z)
    H.add_edge(Z, Y)
    return tuple(added_edges)

# 算法 2
# ==========================================================
"""
    根据 FCGR 算法 2 构造完整图 H。
    返回
    ----------
    H:
        完整转换图。
    n_swaps:
        顺序交换次数。
    n_reversals:
        覆盖反转次数。
    n_cover_additions:
        覆盖反转过程中新增的边数。
"""
def _construct_full_H(G: nx.DiGraph, beta: Sequence[Node], alpha: Sequence[Node]) -> tuple[nx.DiGraph, int, int, int]:
    beta_current = list(beta)
    alpha_list = list(alpha)

    beta_position = _validate_order(G, beta_current, "beta", require_topological=True)
    alpha_position = _validate_order(G, alpha_list, "alpha", require_topological=False)
    H = G.copy()
    n_swaps = 0
    n_reversals = 0
    n_cover_additions = 0
    # 从目标顺序 alpha 的最右端开始处理
    for Y in reversed(alpha_list):
        while beta_position[Y] < len(beta_current) - 1:
            current_position = beta_position[Y]
            Z = beta_current[current_position + 1]
             # Z 在目标顺序中本来就在 Y 的右侧，
            # 此时 Y 已经到达正确位置
            if alpha_position[Z] > alpha_position[Y]:
                break
            if H.has_edge(Y, Z):
                added_edges = _cover_and_reverse(H, Y, Z)
                n_reversals += 1
                n_cover_additions += len(added_edges)
            else:
                added_edges = tuple()
            # 在当前 beta 中交换相邻节点 Y 和 Z
            beta_current[current_position] = Z
            beta_current[current_position + 1] = Y
            beta_position[Z] = current_position
            beta_position[Y] = current_position + 1
            n_swaps += 1

    if beta_current != alpha_list:
        raise AssertionError(f"算法 2 终止后 beta 不等于 alpha：\nbeta={beta_current}\nalpha={alpha_list}")
    for source, target in H.edges:
        #检查 H 的全部边是否与目标顺序 alpha 一致
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError(f"边 {source!r}->{target!r} 与目标顺序 alpha 不一致。")
    if not nx.is_directed_acyclic_graph(H):
        raise AssertionError("算法 2 输出的完整图 H 不是 DAG。")
    H.name = f"{G.name or 'G'}_H"
    return H, n_swaps, n_reversals, n_cover_additions

#统一调用函数并封装结果
"""
    执行完整 FCGR。
    beta:
        显式给定的原图合法拓扑序。
    seed:
        beta=None 时，用于生成可复现随机拓扑序。
    record_trace:
        True 时保存算法 2 的逐步执行轨迹；
        False 时 trace 为空元组。
    无论 record_trace 取何值，
    函数始终返回 FCGRResult。
"""
def run_fcgr(G: nx.DiGraph, C: Iterable[Node], beta: Sequence[Node] | None = None, seed: int | None = None) -> FCGRResult:
    orders = construct_algorithm1_orders(G=G, C=C, beta=beta, seed=seed)# 算法 1
    (H, n_swaps, n_reversals, n_cover_additions) = _construct_full_H(G, orders.beta, orders.alpha)# 算法 2
    # 原图在 R 上的导出子图
    G_R = G.subgraph(orders.alpha_R).copy()
    G_R.name = f"{G.name or 'G'}_GR"
    # FCGR 最终输出 H_R=H[R]
    H_R = H.subgraph(orders.alpha_R).copy()
    H_R.name = f"{G.name or 'G'}_HR"
    # 最终条件图中真正新增的 R 内部边
    added_edges_R = tuple(sorted(set(H_R.edges()) - set(G_R.edges()), key=_edge_sort_key))

    return FCGRResult(
        H=H,
        H_R=H_R,
        G_R=G_R,
        orders=orders,
        added_edges_R=added_edges_R,
        n_swaps=n_swaps,
        n_reversals=n_reversals,
        n_cover_additions=n_cover_additions,
    )
```

## ROS 算法

```python
# 输入验证
def _validate_dag(G: nx.DiGraph) -> None:
    """
    检查 G 是否为简单有向无环图。
    """
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。")
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")

def _validate_conditioning_set(
    G: nx.DiGraph,
    C: Iterable[Node],
) -> set[Node]:
    """
    检查条件集 C 是否为 V 的非空真子集。
    """
    C_set = set(C)
    V_set = set(G.nodes)

    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。")
    return C_set

def _validate_topological_order(
    G: nx.DiGraph,
    alpha: Sequence[Node],
) -> tuple[list[Node], dict[Node, int]]:
    """
    检查 alpha 是否恰好包含 G 的全部节点一次， 并且是 G 的合法拓扑序。
    返回：
        alpha_list：
            alpha 的列表形式。
        alpha_position：
            节点到其顺序位置的映射。
    """
    alpha_list = list(alpha)
    V_set = set(G.nodes)
    if len(alpha_list) != G.number_of_nodes() or set(alpha_list) != V_set:
        raise ValueError("alpha 必须恰好包含 G 中的每个节点一次。")
    alpha_position = {node: index for index, node in enumerate(alpha_list)}
    for source, target in G.edges:
        if alpha_position[source] >= alpha_position[target]:
            raise ValueError("alpha 不是原图 G 的合法拓扑序：" f"边 {source!r}->{target!r} 与该顺序冲突。")
    return alpha_list, alpha_position

#返回 node 的全部祖先，并包含 node 自身。
def ancestors_with_self(
    G: nx.DiGraph,
    node: Node,
) -> set[Node]:
    if node not in G:
        raise ValueError(  f"节点 {node!r} 不在图 G 中。"   )
    ancestors = set(nx.ancestors(G, node))
    ancestors.add(node)
    return ancestors

# ROS 算法

def minimal_fill_edge_set(
    G: nx.DiGraph,
    alpha: Sequence[Node],
    C: Iterable[Node],
) -> dict:
    """
    ROS 算法：寻找条件化模型的最小填充边集。
        E_alpha_R:
            ROS 找到的新增有向边，每条边保存为 (source, target)，
        G_alpha_R:
            G_R 加上 E_alpha_R 后得到的条件模型 minimal I-map。
    """
   
    # 0. 输入验证
    _validate_dag(G)
    C_set = _validate_conditioning_set(G=G, C=C)
    alpha_list, alpha_position = _validate_topological_order(G=G, alpha=alpha)
    # 保存原始图，不修改输入对象
    G_original = G
    G_work = G.copy()
    
    # 1. 构造 G^C：删除 C 中节点的全部出边
    for conditioned_node in C_set:
        outgoing_edges = [(conditioned_node, child) for child in G_work.successors(conditioned_node)]
        G_work.remove_edges_from(outgoing_edges)
    
    # 2. 计算 An_{G^C}(C)
    nodes_of_interest: set[Node] = set()
    for conditioned_node in C_set:
        nodes_of_interest.update(ancestors_with_self(G=G_work, node=conditioned_node))
    
    # 3. 构造祖先闭包上的诱导子图 G_0
    G_0 = G_work.subgraph(nodes_of_interest).copy()
    G_0.name = f"{G.name or 'G'}_ROS_search"
    
    # 4. 按 alpha 的逆序排列搜索节点
    vertices = [node for node in reversed(alpha_list) if node in nodes_of_interest]
    k = len(vertices)

    # 每条新增边保存为真正的节点元组
    E_alpha_R: list[Edge] = []
 
    # 5. 逐层完成父节点集合
    if k >= 3:
        for current_node in vertices[: k - 2]:
            parents = sorted(G_0.predecessors(current_node), key=alpha_position.__getitem__)
            if len(parents) < 2:
                continue
            for x, y in combinations(parents, 2):
                # 已经邻接时不重复添加
                if G_0.has_edge(x, y) or G_0.has_edge(y, x):
                    continue
                # 按 alpha 从左向右定向
                new_edge = (x, y) if alpha_position[x] < alpha_position[y] else (y, x)
                G_0.add_edge(*new_edge)
                E_alpha_R.append(new_edge)

    # 6. 构造原图在 R=V\\C 上的完整导出子图
    R_nodes = [node for node in G_original.nodes if node not in C_set]

    G_alpha_R = G_original.subgraph(R_nodes).copy()
    G_alpha_R.name = f"{G.name or 'G'}_ROS_HR"
    
    # 7. 添加 ROS 找到的填充边
    G_alpha_R.add_edges_from(E_alpha_R)
    
    # 8. 输出检查
    if not nx.is_directed_acyclic_graph(G_alpha_R):
        raise AssertionError("ROS 输出图 G_alpha_R 不是 DAG。")

    for source, target in G_alpha_R.edges:
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError("ROS 输出边与 alpha 不一致：" f"{source!r}->{target!r}。")

    return {"E_alpha_R": E_alpha_R, "G_alpha_R": G_alpha_R}
```

## 标准网络实验框架

```python

# ==========================================================
# 标准网络结构读取
# ==========================================================

EXPECTED_NODE_COUNTS = {
    "Asia": 8,
    "Insurance": 27,
    "Alarm": 37,
    "Win95pts": 76,
    "Andes": 223,
    "Munin": 1041,
}


def _clean_bif_name(raw_name: str) -> str:
    """清理 BIF 文件中的变量名。"""
    name = raw_name.strip()
    if len(name) >= 2 and name[0] == name[-1] and name[0] in {"'", '"'}:
        name = name[1:-1]
    return name.strip()


def _load_bif_structure_fallback(path: Path) -> nx.DiGraph:
    """
    只读取 BIF 中的节点和父子关系。

    当环境中没有安装 pgmpy 时使用该后备解析器。
    本实验只研究图结构，不读取 CPT。
    """
    text = path.read_text(encoding="utf-8", errors="ignore")

    # 删除块注释和行注释
    text = re.sub(r"/\*.*?\*/", "", text, flags=re.S)
    text = re.sub(r"//.*?$", "", text, flags=re.M)

    G = nx.DiGraph(name=path.stem)

    # variable node_name { ... }
    variable_pattern = re.compile(
        r"\bvariable\s+(\"[^\"]+\"|'[^']+'|[^\s\{]+)\s*\{",
        flags=re.I,
    )
    for match in variable_pattern.finditer(text):
        node = _clean_bif_name(match.group(1))
        if node:
            G.add_node(node)

    # probability ( child | parent1, parent2 ) { ... }
    probability_pattern = re.compile(
        r"\bprobability\s*\(\s*(.*?)\s*\)\s*\{",
        flags=re.I | re.S,
    )
    for match in probability_pattern.finditer(text):
        header = match.group(1).strip()

        if "|" in header:
            child_text, parent_text = header.split("|", 1)
            parents = [
                _clean_bif_name(item)
                for item in parent_text.split(",")
                if _clean_bif_name(item)
            ]
        else:
            child_text = header
            parents = []

        child = _clean_bif_name(child_text)
        if not child:
            continue

        G.add_node(child)
        for parent in parents:
            G.add_edge(parent, child)

    if G.number_of_nodes() == 0:
        raise ValueError(f"未能从 BIF 文件读取任何节点：{path}")

    return G


def load_bif_dag(path: str | Path) -> nx.DiGraph:
    """
    从 BIF 文件读取 DAG。

    优先使用 pgmpy；若未安装 pgmpy，则使用只读取结构的后备解析器。
    """
    path = Path(path)

    if not path.exists():
        raise FileNotFoundError(f"BIF 文件不存在：{path}")

    try:
        from pgmpy.readwrite import BIFReader
    except ImportError:
        G = _load_bif_structure_fallback(path)
    else:
        model = BIFReader(str(path)).get_model()
        G = nx.DiGraph(name=path.stem)
        G.add_nodes_from(model.nodes())
        G.add_edges_from(model.edges())

    _validate_dag(G)
    return G


def load_standard_networks(
    data_dir: str | Path,
    network_files: Mapping[str, str],
    strict_node_count: bool = True,
) -> dict[str, nx.DiGraph]:
    """
    读取 Asia、Insurance、Alarm、Win95pts、Andes、Munin。
    """
    data_dir = Path(data_dir)
    graphs: dict[str, nx.DiGraph] = {}

    for network, filename in network_files.items():
        path = data_dir / filename
        G = load_bif_dag(path)
        G.name = network

        expected_n = EXPECTED_NODE_COUNTS.get(network)
        if strict_node_count and expected_n is not None:
            if G.number_of_nodes() != expected_n:
                raise ValueError(
                    f"{network} 节点数不正确："
                    f"读取到 {G.number_of_nodes()}，预期 {expected_n}。"
                    f"请检查是否使用了正确的 BIF 文件。"
                )

        graphs[network] = G

    return graphs


# ==========================================================
# 实验配置
# ==========================================================

@dataclass(frozen=True)
class StandardConsistencyConfig:
    """
    六个标准网络上的 FCGR–ROS 结构一致性实验配置。

    每个实例：
        1. 均匀随机抽取一个大小为 c_size 的条件集 C；
        2. 随机生成原图 G 的一个合法拓扑序 beta；
        3. FCGR 和 ROS 使用完全相同的 (G, C, beta)。
    """
    c_size: int = 2
    cases_per_network: int = 30
    base_seed: int = 20260729
    stop_on_mismatch: bool = False


# ==========================================================
# 结果序列化与随机输入生成
# ==========================================================

def _node_tuple(nodes: Iterable[Node]) -> tuple[Node, ...]:
    return tuple(sorted(nodes, key=str))


def _encode_nodes(nodes: Iterable[Node]) -> str:
    return json.dumps(
        [str(node) for node in _node_tuple(nodes)],
        ensure_ascii=False,
    )


def _encode_order(order: Sequence[Node]) -> str:
    """
    保留拓扑序原始位置，不能使用排序后的节点集合。
    """
    return json.dumps(
        [str(node) for node in order],
        ensure_ascii=False,
    )


def _encode_edges(edges: Iterable[Edge]) -> str:
    ordered = sorted(
        edges,
        key=lambda edge: (str(edge[0]), str(edge[1])),
    )
    return json.dumps(
        [[str(source), str(target)] for source, target in ordered],
        ensure_ascii=False,
    )


def sample_random_condition_sets(
    G: nx.DiGraph,
    n_cases: int,
    c_size: int,
    seed: int,
) -> list[tuple[Node, ...]]:
    """
    均匀随机抽取条件集。

    在组合数允许时优先保证条件集互不重复；
    如果 n_cases 大于全部可能条件集数量，则在枚举完全部不同条件集后重复抽取。
    """
    nodes = list(G.nodes())

    if not 1 <= c_size < len(nodes):
        raise ValueError("c_size 必须满足 1 <= c_size < |V|。")

    if n_cases <= 0:
        raise ValueError("n_cases 必须为正整数。")

    rng = random.Random(seed)

    max_unique = math.comb(len(nodes), c_size)
    unique_target = min(n_cases, max_unique)

    condition_sets: list[tuple[Node, ...]] = []
    seen: set[tuple[Node, ...]] = set()

    max_attempts = max(1000, 50 * unique_target)
    attempts = 0

    while len(condition_sets) < unique_target and attempts < max_attempts:
        attempts += 1

        C = tuple(
            sorted(
                rng.sample(nodes, c_size),
                key=str,
            )
        )

        if C not in seen:
            seen.add(C)
            condition_sets.append(C)

    if len(condition_sets) != unique_target:
        raise RuntimeError("无法生成足够数量的互不重复条件集。")

    unique_sets = list(condition_sets)

    while len(condition_sets) < n_cases:
        condition_sets.append(rng.choice(unique_sets))

    rng.shuffle(condition_sets)
    return condition_sets


def ancestor_closure_after_cut(
    G: nx.DiGraph,
    C: Iterable[Node],
) -> set[Node]:
    """
    计算 ROS 实际搜索的 An_{G^C}(C)。

    G^C 表示删除条件节点全部出边后的图。
    """
    C_set = set(C)
    G_cut = G.copy()

    for conditioned_node in C_set:
        G_cut.remove_edges_from(
            list(G_cut.out_edges(conditioned_node))
        )

    closure: set[Node] = set()

    for conditioned_node in C_set:
        closure.update(nx.ancestors(G_cut, conditioned_node))
        closure.add(conditioned_node)

    return closure


def _all_edges_follow_order(
    edges: Iterable[Edge],
    order_position: Mapping[Node, int],
) -> bool:
    return all(
        order_position[source] < order_position[target]
        for source, target in edges
    )


# ==========================================================
# 单个配对实例
# ==========================================================

def run_one_consistency_case(
    network: str,
    G: nx.DiGraph,
    case_id: int,
    C: Sequence[Node],
    beta: Sequence[Node],
    condition_seed: int,
    beta_seed: int,
) -> dict:
    """
    在同一个 (G, C, beta) 下运行 FCGR 和 ROS，并比较结构。
    """
    C_set = set(C)
    R_set = set(G.nodes()) - C_set

    G_R = G.subgraph(R_set).copy()
    base_edges = set(G_R.edges())

    beta_position = {
        node: index
        for index, node in enumerate(beta)
    }

    # 分别复制输入，避免任一算法影响另一算法
    G_fcgr = G.copy()
    G_ros = G.copy()

    fcgr_before = (
        set(G_fcgr.nodes()),
        set(G_fcgr.edges()),
    )
    ros_before = (
        set(G_ros.nodes()),
        set(G_ros.edges()),
    )

    row = {
        "network": network,
        "case_id": case_id,
        "condition_seed": condition_seed,
        "beta_seed": beta_seed,
        "n_nodes": G.number_of_nodes(),
        "n_edges": G.number_of_edges(),
        "c_size": len(C_set),
        "r_size": len(R_set),
        "C": _encode_nodes(C_set),
        "beta": _encode_order(beta),
        "ancestor_closure_size": len(
            ancestor_closure_after_cut(G, C_set)
        ),
        "status": "ok",
        "error_type": "",
        "error_message": "",
    }

    try:
        # 两个算法使用相同的 G、C 和基础拓扑序 beta
        fcgr_result = run_fcgr(
            G=G_fcgr,
            C=C_set,
            beta=beta,
        )

        ros_result = minimal_fill_edge_set(
            G=G_ros,
            alpha=beta,
            C=C_set,
        )

        H_fcgr = fcgr_result.H_R
        H_ros = ros_result["G_alpha_R"]

        fcgr_edges = set(H_fcgr.edges())
        ros_edges = set(H_ros.edges())

        fcgr_fill = set(fcgr_result.added_edges_R)
        ros_fill = set(ros_result["E_alpha_R"])

        node_equal = (
            set(H_fcgr.nodes())
            == set(H_ros.nodes())
            == R_set
        )

        graph_equal = (
            node_equal
            and fcgr_edges == ros_edges
        )

        fill_equal = (
            fcgr_fill == ros_fill
        )

        fill_count_equal = (
            len(fcgr_fill) == len(ros_fill)
        )

        parent_equal = (
            node_equal
            and all(
                set(H_fcgr.predecessors(node))
                == set(H_ros.predecessors(node))
                for node in R_set
            )
        )

        fcgr_is_dag = nx.is_directed_acyclic_graph(H_fcgr)
        ros_is_dag = nx.is_directed_acyclic_graph(H_ros)

        fcgr_preserves_base = base_edges.issubset(fcgr_edges)
        ros_preserves_base = base_edges.issubset(ros_edges)

        fcgr_order_valid = _all_edges_follow_order(
            fcgr_edges,
            beta_position,
        )
        ros_order_valid = _all_edges_follow_order(
            ros_edges,
            beta_position,
        )

        fcgr_input_unchanged = (
            fcgr_before
            == (
                set(G_fcgr.nodes()),
                set(G_fcgr.edges()),
            )
        )
        ros_input_unchanged = (
            ros_before
            == (
                set(G_ros.nodes()),
                set(G_ros.edges()),
            )
        )

        only_fcgr = fcgr_fill - ros_fill
        only_ros = ros_fill - fcgr_fill

        all_checks_pass = all([
            node_equal,
            graph_equal,
            fill_equal,
            fill_count_equal,
            parent_equal,
            fcgr_is_dag,
            ros_is_dag,
            fcgr_preserves_base,
            ros_preserves_base,
            fcgr_order_valid,
            ros_order_valid,
            fcgr_input_unchanged,
            ros_input_unchanged,
        ])

        row.update({
            "alpha": _encode_order(fcgr_result.orders.alpha),
            "alpha_R": _encode_order(fcgr_result.orders.alpha_R),
            "G_R_edge_count": len(base_edges),
            "FCGR_edge_count": len(fcgr_edges),
            "ROS_edge_count": len(ros_edges),
            "FCGR_fill_count": len(fcgr_fill),
            "ROS_fill_count": len(ros_fill),
            "node_equal": node_equal,
            "graph_equal": graph_equal,
            "fill_equal": fill_equal,
            "fill_count_equal": fill_count_equal,
            "parent_equal": parent_equal,
            "FCGR_is_dag": fcgr_is_dag,
            "ROS_is_dag": ros_is_dag,
            "FCGR_preserves_G_R": fcgr_preserves_base,
            "ROS_preserves_G_R": ros_preserves_base,
            "FCGR_order_valid": fcgr_order_valid,
            "ROS_order_valid": ros_order_valid,
            "FCGR_input_unchanged": fcgr_input_unchanged,
            "ROS_input_unchanged": ros_input_unchanged,
            "all_checks_pass": all_checks_pass,
            "FCGR_fill_edges": _encode_edges(fcgr_fill),
            "ROS_fill_edges": _encode_edges(ros_fill),
            "only_FCGR_edges": _encode_edges(only_fcgr),
            "only_ROS_edges": _encode_edges(only_ros),
        })

    except Exception as exc:
        row.update({
            "status": "error",
            "error_type": type(exc).__name__,
            "error_message": str(exc),
            "all_checks_pass": False,
        })

    return row


# ==========================================================
# 六个标准网络的完整实验
# ==========================================================

def run_standard_consistency_experiment(
    network_graphs: Mapping[str, nx.DiGraph],
    config: StandardConsistencyConfig = StandardConsistencyConfig(),
) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    对六个网络分别运行随机条件集、随机拓扑序的配对一致性实验。
    """
    rows: list[dict] = []

    for network_index, (network, G) in enumerate(
        network_graphs.items(),
        start=1,
    ):
        _validate_dag(G)

        network_seed = (
            config.base_seed
            + network_index * 1_000_000
        )

        condition_seed = network_seed + 101

        condition_sets = sample_random_condition_sets(
            G=G,
            n_cases=config.cases_per_network,
            c_size=config.c_size,
            seed=condition_seed,
        )

        for case_id, C in enumerate(
            condition_sets,
            start=1,
        ):
            beta_seed = (
                network_seed
                + 10_000
                + case_id
            )

            beta = random_topological_order(
                G=G,
                seed=beta_seed,
            )

            row = run_one_consistency_case(
                network=network,
                G=G,
                case_id=case_id,
                C=C,
                beta=beta,
                condition_seed=condition_seed,
                beta_seed=beta_seed,
            )

            rows.append(row)

            if (
                config.stop_on_mismatch
                and not row["all_checks_pass"]
            ):
                raise AssertionError(
                    f"{network} case {case_id} "
                    "未通过一致性检查。"
                )

    details = pd.DataFrame(rows)

    mismatch_mask = (
        (details["status"] != "ok")
        | (~details["all_checks_pass"].fillna(False))
    )
    mismatches = details.loc[mismatch_mask].copy()

    summary_rows: list[dict] = []

    for network, group in details.groupby(
        "network",
        sort=False,
    ):
        successful = group[group["status"] == "ok"]

        n_total = len(group)
        n_successful = len(successful)

        def rate_percent(column: str) -> float:
            if n_successful == 0:
                return float("nan")
            return 100.0 * float(successful[column].mean())

        summary_rows.append({
            "network": network,
            "n_cases": n_total,
            "n_successful": n_successful,
            "n_errors": n_total - n_successful,
            "unique_condition_sets": group["C"].nunique(),
            "graph_equal_rate_pct": rate_percent("graph_equal"),
            "fill_equal_rate_pct": rate_percent("fill_equal"),
            "parent_equal_rate_pct": rate_percent("parent_equal"),
            "all_checks_pass_rate_pct": rate_percent(
                "all_checks_pass"
            ),
            "nonzero_fill_cases": (
                int(
                    (
                        successful["FCGR_fill_count"] > 0
                    ).sum()
                )
                if n_successful
                else 0
            ),
            "mean_fill_count": (
                float(
                    successful["FCGR_fill_count"].mean()
                )
                if n_successful
                else float("nan")
            ),
            "max_fill_count": (
                int(
                    successful["FCGR_fill_count"].max()
                )
                if n_successful
                else 0
            ),
        })

    summary = pd.DataFrame(summary_rows)

    return details, summary, mismatches


# ==========================================================
# 保存结果
# ==========================================================

def save_consistency_results(
    details: pd.DataFrame,
    summary: pd.DataFrame,
    mismatches: pd.DataFrame,
    config: StandardConsistencyConfig,
    output_file: str | Path,
) -> Path:
    output_file = Path(output_file)
    output_file.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    config_df = pd.DataFrame({
        "parameter": [
            "c_size",
            "cases_per_network",
            "base_seed",
            "stop_on_mismatch",
        ],
        "value": [
            config.c_size,
            config.cases_per_network,
            config.base_seed,
            config.stop_on_mismatch,
        ],
    })

    with pd.ExcelWriter(
        output_file,
        engine="openpyxl",
    ) as writer:
        summary.to_excel(
            writer,
            sheet_name="network_summary",
            index=False,
        )
        details.to_excel(
            writer,
            sheet_name="instance_results",
            index=False,
        )
        mismatches.to_excel(
            writer,
            sheet_name="mismatches",
            index=False,
        )
        config_df.to_excel(
            writer,
            sheet_name="config",
            index=False,
        )

    return output_file
```

## 结果判定

主要判定指标：

- `fill_equal`：FCGR 与 ROS 的填充边集合完全相同；
- `graph_equal`：两个算法的最终输出图节点集与有向边集完全相同；
- `parent_equal`：对每个剩余节点，两图父节点集合完全相同；
- `all_checks_pass`：同时通过 DAG、原边保持、顺序方向和输入未修改等检查。

`mismatches` 工作表会保存全部不一致或报错实例，用于复现和定位问题。


```python

# ==========================================================
# 配置网络文件与实验参数
# ==========================================================

# 修改成六个 BIF 文件所在目录
DATA_DIR = Path(r"D:\your_project\data")

NETWORK_FILES = {
    "Asia": "asia.bif",
    "Insurance": "insurance.bif",
    "Alarm": "alarm.bif",
    "Win95pts": "win95pts.bif",
    "Andes": "andes.bif",
    "Munin": "munin.bif",
}

# 初次调试可设为 5；正式结构一致性实验建议设为 30 或 100
CONFIG = StandardConsistencyConfig(
    c_size=2,
    cases_per_network=30,
    base_seed=20260729,
    stop_on_mismatch=False,
)

RESULT_FILE = Path("./results/FCGR_ROS_standard_consistency.xlsx")
```

```python

# ==========================================================
# 运行实验
# ==========================================================

network_graphs = load_standard_networks(
    data_dir=DATA_DIR,
    network_files=NETWORK_FILES,
    strict_node_count=True,
)

network_info = pd.DataFrame([
    {
        "network": name,
        "n_nodes": graph.number_of_nodes(),
        "n_edges": graph.number_of_edges(),
        "is_dag": nx.is_directed_acyclic_graph(graph),
    }
    for name, graph in network_graphs.items()
])

display(network_info)

details, summary, mismatches = (
    run_standard_consistency_experiment(
        network_graphs=network_graphs,
        config=CONFIG,
    )
)

display(summary)

print(f"总实例数：{len(details)}")
print(f"不一致或报错实例数：{len(mismatches)}")

if len(mismatches) > 0:
    display(
        mismatches[[
            "network",
            "case_id",
            "C",
            "beta_seed",
            "status",
            "error_type",
            "graph_equal",
            "fill_equal",
            "only_FCGR_edges",
            "only_ROS_edges",
        ]]
    )
else:
    print("全部实例的填充边集、输出图和逐节点父集均一致。")
```

```python

# ==========================================================
# 保存 Excel
# ==========================================================

saved_path = save_consistency_results(
    details=details,
    summary=summary,
    mismatches=mismatches,
    config=CONFIG,
    output_file=RESULT_FILE,
)

print(f"结果已保存到：{saved_path.resolve()}")
```
