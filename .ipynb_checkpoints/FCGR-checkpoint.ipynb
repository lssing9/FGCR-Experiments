```python
import random
from dataclasses import dataclass
from typing import Hashable, Iterable, Sequence
import networkx as nx
Node = Hashable
Edge = tuple[Node, Node]
```

```python
# 算法 1 的输出
@dataclass(frozen=True)
class Algorithm1Orders:
    """
    FCGR 算法 1 产生的全部顺序。
    """
    beta: tuple[Node, ...]
    alpha_C: tuple[Node, ...]
    alpha_R: tuple[Node, ...]
    alpha: tuple[Node, ...]
    seed: int | None = None

# FCGR 统一结果对象
@dataclass
class FCGRResult:
    H: nx.DiGraph
    H_R: nx.DiGraph
    G_R: nx.DiGraph
    orders: Algorithm1Orders
    # 最终 H_R 相对于 G_R 新增的边
    added_edges_R: tuple[Edge, ...]
    # 算法 2 运行统计
    n_swaps: int
    n_reversals: int
    n_cover_additions: int


# 通用输入检查


def _validate_dag(G: nx.DiGraph) -> None:
    """
    检查 G 是否为简单 DAG。
    """
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。"  )
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")


def _validate_conditioning_set(G: nx.DiGraph, C: Iterable[Node]) -> set[Node]:
    """
    检查 C 是否为 V 的非空真子集。
    """
    C_set = set(C)
    V_set = set(G.nodes)
    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。" )
    return C_set


def _validate_order(G: nx.DiGraph, order: Sequence[Node], name: str, require_topological: bool = False) -> dict[Node, int]:
    """
    检查 order 是否恰好包含 V 中全部节点一次。
    require_topological=True 时，
    还要求 order 是 G 的合法拓扑序。
    返回节点到顺序位置的映射。
    """
    order_list = list(order)
    V_set = set(G.nodes)
    if len(order_list) != G.number_of_nodes() or set(order_list) != V_set:
        raise ValueError(f"{name} 必须恰好包含 V 中的每个节点一次。")
    position = {node: index for index, node in enumerate(order_list)}
    if require_topological:
        for source, target in G.edges:
            if position[source] >= position[target]:
                raise ValueError(  f"{name} 不是 G 的合法拓扑序：" f"边 {source!r}->{target!r} " "与该顺序冲突。" )
    return position


def _edge_sort_key(
    edge: Edge,
) -> tuple[str, str]:
    """
    支持字符串、整数等异构节点的稳定边排序。
    """
    return str(edge[0]), str(edge[1])        

# ==========================================================
# 随机拓扑序
# ==========================================================
def random_topological_order(G: nx.DiGraph, seed: int | None = None) -> list[Node]:
    """
    生成 G 的一个合法拓扑序。
    seed=None：
        每次选择按字符串排序后的第一个零入度节点，
        因此结果确定。
    seed为整数：
        在当前零入度节点中随机选择，且结果可复现。
    """
    _validate_dag(G)
    rng = random.Random(seed)
    indegree = dict(G.in_degree())
    available = sorted([node for node, degree in indegree.items() if degree == 0], key=str)
    beta: list[Node] = []
    while available:
        selected_index = 0 if seed is None else rng.randrange(len(available))
        selected_node = available.pop(selected_index)
        beta.append(selected_node)
        for child in sorted(G.successors(selected_node), key=str):
            indegree[child] -= 1
            if indegree[child] == 0:
                available.append(child)
        available.sort(key=str)
    if len(beta) != G.number_of_nodes():
        raise RuntimeError("拓扑排序未覆盖 G 的全部节点。")
    return beta


# ==========================================================
# 算法 1：构造 beta 与 alpha
"""
    构造 FCGR 所需顺序。
    两种调用方式
    ----------
    1. 显式给定 beta：
       construct_algorithm1_orders(
           G=G,
           C=C,
           beta=beta,
       )
    2. 根据 seed 生成 beta：
       construct_algorithm1_orders(
           G=G,
           C=C,
           seed=1,
       )
    beta 和 seed 不能同时给定。
"""
def construct_algorithm1_orders(G: nx.DiGraph, C: Iterable[Node], beta: Sequence[Node] | None = None, seed: int | None = None) -> Algorithm1Orders:
    _validate_dag(G)
    C_set = _validate_conditioning_set(G, C)
    if beta is not None and seed is not None:
        raise ValueError("beta 和 seed 不能同时指定。")
    if beta is None:
        beta_list = random_topological_order(G, seed=seed)
        result_seed = seed
    else:
        beta_list = list(beta)
        _validate_order(G, beta_list, "beta", require_topological=True)
        result_seed = None

    alpha_C = [node for node in beta_list if node in C_set]
    alpha_R = [node for node in beta_list if node not in C_set]
    alpha = alpha_C + alpha_R

    return Algorithm1Orders(
        beta=tuple(beta_list),
        alpha_C=tuple(alpha_C),
        alpha_R=tuple(alpha_R),
        alpha=tuple(alpha),
        seed=result_seed,
    )


# 覆盖并反转 Y→Z
"""
    将当前边 Y→Z 补成 covered edge，
    然后反转为 Z→Y。
    返回覆盖过程中新增的边，
    不包含反转后的边 Z→Y。
"""
# ==========================================================
def _cover_and_reverse(H: nx.DiGraph, Y: Node, Z: Node) -> tuple[Edge, ...]:
    if not H.has_edge(Y, Z):
        raise ValueError(f"当前图中不存在边 {Y!r}->{Z!r}。")
    pa_Y = set(H.predecessors(Y))
    pa_Z = set(H.predecessors(Z))
    added_edges: list[Edge] = []
    # 将 pa(Y)\\pa(Z) 补成 Z 的父节点
    for X in sorted(pa_Y - pa_Z, key=str):
        if not H.has_edge(X, Z):
            H.add_edge(X, Z)
            added_edges.append((X, Z))
    # 将 pa(Z)\\(pa(Y)∪{Y}) 补成 Y 的父节点
    for X in sorted(pa_Z - pa_Y - {Y}, key=str):
        if not H.has_edge(X, Y):
            H.add_edge(X, Y)
            added_edges.append((X, Y))
    # 覆盖边条件：
    # pa(Z) = pa(Y) ∪ {Y}
    if set(H.predecessors(Z)) != set(H.predecessors(Y)) | {Y}:
        raise AssertionError(f"边 {Y!r}->{Z!r} 在反转前未成为 covered edge。")

    H.remove_edge(Y, Z)
    H.add_edge(Z, Y)
    return tuple(added_edges)

# 算法 2
# ==========================================================
"""
    根据 FCGR 算法 2 构造完整图 H。
    返回
    ----------
    H:
        完整转换图。
    n_swaps:
        顺序交换次数。
    n_reversals:
        覆盖反转次数。
    n_cover_additions:
        覆盖反转过程中新增的边数。
"""
def _construct_full_H(G: nx.DiGraph, beta: Sequence[Node], alpha: Sequence[Node]) -> tuple[nx.DiGraph, int, int, int]:
    beta_current = list(beta)
    alpha_list = list(alpha)

    beta_position = _validate_order(G, beta_current, "beta", require_topological=True)
    alpha_position = _validate_order(G, alpha_list, "alpha", require_topological=False)
    H = G.copy()
    n_swaps = 0
    n_reversals = 0
    n_cover_additions = 0
    # 从目标顺序 alpha 的最右端开始处理
    for Y in reversed(alpha_list):
        while beta_position[Y] < len(beta_current) - 1:
            current_position = beta_position[Y]
            Z = beta_current[current_position + 1]
             # Z 在目标顺序中本来就在 Y 的右侧，
            # 此时 Y 已经到达正确位置
            if alpha_position[Z] > alpha_position[Y]:
                break
            if H.has_edge(Y, Z):
                added_edges = _cover_and_reverse(H, Y, Z)
                n_reversals += 1
                n_cover_additions += len(added_edges)
            else:
                added_edges = tuple()
            # 在当前 beta 中交换相邻节点 Y 和 Z
            beta_current[current_position] = Z
            beta_current[current_position + 1] = Y
            beta_position[Z] = current_position
            beta_position[Y] = current_position + 1
            n_swaps += 1

    if beta_current != alpha_list:
        raise AssertionError(f"算法 2 终止后 beta 不等于 alpha：\nbeta={beta_current}\nalpha={alpha_list}")
    for source, target in H.edges:
        #检查 H 的全部边是否与目标顺序 alpha 一致
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError(f"边 {source!r}->{target!r} 与目标顺序 alpha 不一致。")
    if not nx.is_directed_acyclic_graph(H):
        raise AssertionError("算法 2 输出的完整图 H 不是 DAG。")
    H.name = f"{G.name or 'G'}_H"
    return H, n_swaps, n_reversals, n_cover_additions

#统一调用函数并封装结果
"""
    执行完整 FCGR。
    beta:
        显式给定的原图合法拓扑序。
    seed:
        beta=None 时，用于生成可复现随机拓扑序。
    record_trace:
        True 时保存算法 2 的逐步执行轨迹；
        False 时 trace 为空元组。
    无论 record_trace 取何值，
    函数始终返回 FCGRResult。
"""
def run_fcgr(G: nx.DiGraph, C: Iterable[Node], beta: Sequence[Node] | None = None, seed: int | None = None) -> FCGRResult:
    orders = construct_algorithm1_orders(G=G, C=C, beta=beta, seed=seed)# 算法 1
    (H, n_swaps, n_reversals, n_cover_additions) = _construct_full_H(G, orders.beta, orders.alpha)# 算法 2
    # 原图在 R 上的导出子图
    G_R = G.subgraph(orders.alpha_R).copy()
    G_R.name = f"{G.name or 'G'}_GR"
    # FCGR 最终输出 H_R=H[R]
    H_R = H.subgraph(orders.alpha_R).copy()
    H_R.name = f"{G.name or 'G'}_HR"
    # 最终条件图中真正新增的 R 内部边
    added_edges_R = tuple(sorted(set(H_R.edges()) - set(G_R.edges()), key=_edge_sort_key))

    return FCGRResult(
        H=H,
        H_R=H_R,
        G_R=G_R,
        orders=orders,
        added_edges_R=added_edges_R,
        n_swaps=n_swaps,
        n_reversals=n_reversals,
        n_cover_additions=n_cover_additions,
    )
```

```python
G = nx.DiGraph(name="test_1")
G.add_nodes_from([  "a", "b", "c", "d", "e"])
G.add_edges_from([ ("a", "c"), ("b", "d"), ("c", "e"), ("d", "e"),])
C = {"e"}

beta = [ "a", "b", "c", "d", "e",]

fcgr_result = run_fcgr( G=G, C=C, beta=beta, )

print("FCGR 新增边：")
print(fcgr_result.added_edges_R)
```

```python
#根据随机种子生成 β
fcgr_result = run_fcgr(
    G=G,
    C=C,
    seed=1,
)

print(fcgr_result.orders.beta)
print(fcgr_result.H_R.edges())
```
