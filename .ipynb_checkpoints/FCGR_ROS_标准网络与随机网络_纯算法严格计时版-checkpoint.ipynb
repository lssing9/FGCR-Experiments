# FCGR 与 ROS：标准网络和随机网络独立实验合集（纯算法严格计时版）

四项实验仍分别执行、分别保存；公共算法只定义一次。

正式计时边界：

- FCGR 算法1：仅构造 `alpha=(alpha_C, alpha_R)`；
- FCGR 算法2：仅复制原图、执行相邻交换与覆盖反转、提取 `H_R=H[R]`；
- ROS：仅执行最小填充边集搜索，生成原始填充边列表后立即停止计时；
- `G_R`、最终填充边差集、ROS 最终图、验证、比较、统计和保存均在计时外。


## 公共依赖与 FCGR/ROS 算法

```python
from __future__ import annotations

import gc
import gzip
import hashlib
import json
import math
import platform
import random
import re
import sys
import time
from dataclasses import asdict, dataclass
from functools import lru_cache
from itertools import combinations
from pathlib import Path
from statistics import median
from typing import Hashable, Iterable, Iterator, Mapping, Sequence

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd


Node = Hashable
Edge = tuple[Node, Node]


# ==========================================================
# 1. 通用检查、编码与哈希
# ==========================================================


def _validate_dag(G: nx.DiGraph) -> None:
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。")
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")


def _validate_conditioning_set(
    G: nx.DiGraph,
    C: Iterable[Node],
) -> set[Node]:
    C_set = set(C)
    V_set = set(G.nodes)
    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。")
    return C_set


def _validate_order(
    G: nx.DiGraph,
    order: Sequence[Node],
    name: str,
    require_topological: bool,
) -> dict[Node, int]:
    order_list = list(order)
    if len(order_list) != G.number_of_nodes():
        raise ValueError(f"{name} 的长度不等于 G 的节点数。")
    if len(set(order_list)) != len(order_list):
        raise ValueError(f"{name} 中存在重复节点。")
    if set(order_list) != set(G.nodes):
        raise ValueError(f"{name} 必须恰好包含 G 的全部节点。")

    position = {node: index for index, node in enumerate(order_list)}
    if require_topological:
        for source, target in G.edges:
            if position[source] >= position[target]:
                raise ValueError(
                    f"{name} 不是 G 的合法拓扑序："
                    f"{source!r}->{target!r} 与顺序冲突。"
                )
    return position


def _edge_sort_key(edge: Edge) -> tuple[str, str]:
    return str(edge[0]), str(edge[1])


def _encode_nodes(nodes: Iterable[Node]) -> str:
    return json.dumps(
        [str(node) for node in sorted(nodes, key=str)],
        ensure_ascii=False,
        separators=(",", ":"),
    )


def _encode_order(order: Sequence[Node]) -> str:
    return json.dumps(
        [str(node) for node in order],
        ensure_ascii=False,
        separators=(",", ":"),
    )


def _encode_edges(edges: Iterable[Edge]) -> str:
    ordered = sorted(edges, key=_edge_sort_key)
    return json.dumps(
        [[str(source), str(target)] for source, target in ordered],
        ensure_ascii=False,
        separators=(",", ":"),
    )


def _sha256_text(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def _hash_nodes(nodes: Iterable[Node]) -> str:
    return _sha256_text(_encode_nodes(nodes))


def _hash_order(order: Sequence[Node]) -> str:
    return _sha256_text(_encode_order(order))


def _hash_edges(edges: Iterable[Edge]) -> str:
    return _sha256_text(_encode_edges(edges))


def _graph_fingerprint(G: nx.DiGraph) -> str:
    payload = {
        "nodes": [str(node) for node in sorted(G.nodes, key=str)],
        "edges": [
            [str(source), str(target)]
            for source, target in sorted(G.edges, key=_edge_sort_key)
        ],
    }
    return _sha256_text(
        json.dumps(
            payload,
            ensure_ascii=False,
            separators=(",", ":"),
        )
    )


def _derive_seed(base_seed: int, *parts: object) -> int:
    """从根种子与实验标识稳定派生 64 位随机种子。"""
    payload = "|".join([str(base_seed), *(str(part) for part in parts)])
    digest = hashlib.blake2b(
        payload.encode("utf-8"),
        digest_size=8,
    ).digest()
    return int.from_bytes(digest, byteorder="big", signed=False)


# ==========================================================
# 2. FCGR 算法1：只构造目标顺序 alpha
# ==========================================================


@dataclass(frozen=True)
class Algorithm1Orders:
    beta: tuple[Node, ...]
    alpha_C: tuple[Node, ...]
    alpha_R: tuple[Node, ...]
    alpha: tuple[Node, ...]


@dataclass
class FCGRResult:
    H: nx.DiGraph
    H_R: nx.DiGraph
    orders: Algorithm1Orders


def random_topological_order(
    G: nx.DiGraph,
    seed: int,
) -> tuple[Node, ...]:
    """随机化Kahn算法，用于在实验开始前生成固定输入beta。"""
    _validate_dag(G)
    rng = random.Random(seed)
    indegree = dict(G.in_degree())
    available = sorted(
        [node for node, degree in indegree.items() if degree == 0],
        key=str,
    )
    order: list[Node] = []

    while available:
        selected_index = rng.randrange(len(available))
        selected_node = available.pop(selected_index)
        order.append(selected_node)

        for child in sorted(G.successors(selected_node), key=str):
            indegree[child] -= 1
            if indegree[child] == 0:
                available.append(child)
        available.sort(key=str)

    if len(order) != G.number_of_nodes():
        raise RuntimeError("拓扑排序未覆盖G的全部节点。")
    return tuple(order)



def construct_algorithm1_orders(
    C_set: set[Node],
    beta: tuple[Node, ...],
) -> Algorithm1Orders:
    """
    FCGR算法1：只构造alpha=(alpha_C, alpha_R)。

    C_set与beta在进入计时区间前已经标准化；本函数不再复制输入，
    不检查G、C或beta。
    """
    alpha_C = tuple(node for node in beta if node in C_set)
    alpha_R = tuple(node for node in beta if node not in C_set)
    alpha = alpha_C + alpha_R
    return Algorithm1Orders(
        beta=beta,
        alpha_C=alpha_C,
        alpha_R=alpha_R,
        alpha=alpha,
    )


def _cover_and_reverse(
    H: nx.DiGraph,
    Y: Node,
    Z: Node,
) -> None:
    """将当前边Y->Z覆盖化并反转。"""
    pa_Y = set(H.predecessors(Y))
    pa_Z = set(H.predecessors(Z))

    for X in sorted(pa_Y - pa_Z, key=str):
        if not H.has_edge(X, Z):
            H.add_edge(X, Z)

    for X in sorted(pa_Z - pa_Y - {Y}, key=str):
        if not H.has_edge(X, Y):
            H.add_edge(X, Y)

    H.remove_edge(Y, Z)
    H.add_edge(Z, Y)


def construct_algorithm2_H_R(
    G: nx.DiGraph,
    orders: Algorithm1Orders,
) -> tuple[nx.DiGraph, nx.DiGraph]:
    """
    FCGR算法2：通过相邻交换和覆盖反转构造H，再提取H_R=H[R]。

    本函数不执行输入验证、covered-edge断言、DAG检查、顺序检查，
    也不构造G_R或计算最终填充边集。
    """
    beta_current = list(orders.beta)
    alpha_list = list(orders.alpha)
    beta_position = {
        node: index
        for index, node in enumerate(beta_current)
    }
    alpha_position = {
        node: index
        for index, node in enumerate(alpha_list)
    }
    H = G.copy()

    for Y in reversed(alpha_list):
        while beta_position[Y] < len(beta_current) - 1:
            current_position = beta_position[Y]
            Z = beta_current[current_position + 1]

            if alpha_position[Z] > alpha_position[Y]:
                break

            if H.has_edge(Y, Z):
                _cover_and_reverse(H, Y, Z)

            beta_current[current_position] = Z
            beta_current[current_position + 1] = Y
            beta_position[Z] = current_position
            beta_position[Y] = current_position + 1

    H_R = H.subgraph(orders.alpha_R).copy()
    return H, H_R



def run_fcgr(
    G: nx.DiGraph,
    C: Iterable[Node],
    beta: Sequence[Node],
) -> FCGRResult:
    """非计时地执行纯FCGR，用于输出一致性实验。"""
    C_set = set(C)
    beta_tuple = tuple(beta)
    orders = construct_algorithm1_orders(
        C_set=C_set,
        beta=beta_tuple,
    )
    H, H_R = construct_algorithm2_H_R(G=G, orders=orders)
    H.name = f"{G.name or 'G'}_H"
    H_R.name = f"{G.name or 'G'}_HR"
    return FCGRResult(
        H=H,
        H_R=H_R,
        orders=orders,
    )


def _timed_fcgr_once(
    G: nx.DiGraph,
    C_set: set[Node],
    beta: tuple[Node, ...],
) -> tuple[FCGRResult, int, int]:
    """
    单次FCGR纯算法计时，单位为纳秒。

    算法1时间只含alpha构造；
    算法2时间只含覆盖反转、完整H形成和H_R提取。
    FCGR最终填充边集在本函数返回后另行计算。
    """
    algorithm1_start = time.perf_counter_ns()
    orders = construct_algorithm1_orders(C_set=C_set, beta=beta)
    algorithm1_end = time.perf_counter_ns()

    algorithm2_start = time.perf_counter_ns()
    H, H_R = construct_algorithm2_H_R(G=G, orders=orders)
    algorithm2_end = time.perf_counter_ns()

    H.name = f"{G.name or 'G'}_H"
    H_R.name = f"{G.name or 'G'}_HR"
    result = FCGRResult(
        H=H,
        H_R=H_R,
        orders=orders,
    )
    return (
        result,
        algorithm1_end - algorithm1_start,
        algorithm2_end - algorithm2_start,
    )


# ==========================================================
# 4. ROS：只保留算法主体
# ==========================================================



def minimal_fill_edge_set(
    G: nx.DiGraph,
    alpha: tuple[Node, ...],
    C_set: set[Node],
) -> list[Edge]:
    """
    ROS算法主体：返回算法直接生成的最小填充边列表。

    本函数不执行输入/输出验证，也不构造G_R、最终输出图，
    不做set转换、排序、去重或结果字典封装。
    """
    alpha_position = {
        node: index
        for index, node in enumerate(alpha)
    }

    # Step 1: 构造G^C。
    G_work = G.copy()
    for conditioned_node in C_set:
        G_work.remove_edges_from(
            list(G_work.out_edges(conditioned_node))
        )

    # Step 2: 计算An_{G^C}(C)并构造G_0。
    nodes_of_interest: set[Node] = set()
    for conditioned_node in C_set:
        nodes_of_interest.update(
            nx.ancestors(G_work, conditioned_node)
        )
        nodes_of_interest.add(conditioned_node)
    G_0 = G_work.subgraph(nodes_of_interest).copy()

    # Step 3: 按alpha逆序逐层完备父集。
    vertices = [
        node
        for node in reversed(alpha)
        if node in nodes_of_interest
    ]
    E_alpha_R: list[Edge] = []

    if len(vertices) >= 3:
        for current_node in vertices[: len(vertices) - 2]:
            parents = sorted(
                G_0.predecessors(current_node),
                key=alpha_position.__getitem__,
            )
            if len(parents) >= 2:
                for x, y in combinations(parents, 2):
                    if G_0.has_edge(x, y) or G_0.has_edge(y, x):
                        continue
                    new_edge = (
                        (x, y)
                        if alpha_position[x] < alpha_position[y]
                        else (y, x)
                    )
                    G_0.add_edge(*new_edge)
                    E_alpha_R.append(new_edge)
            G_0.remove_node(current_node)

    # ROS算法的直接输出；不在这里构造G_R或最终图。
    return E_alpha_R


def _build_ros_result_after_algorithm(
    G: nx.DiGraph,
    C: Iterable[Node],
    E_alpha_R: Iterable[Edge],
) -> dict[str, object]:
    """计时外构造ROS最终图，仅供输出一致性实验使用。"""
    C_set = set(C)
    R_nodes = [
        node
        for node in G.nodes
        if node not in C_set
    ]
    G_alpha_R = G.subgraph(R_nodes).copy()
    fill_edges = tuple(E_alpha_R)
    G_alpha_R.add_edges_from(fill_edges)
    return {
        "E_alpha_R": fill_edges,
        "G_alpha_R": G_alpha_R,
    }


def run_ros(
    G: nx.DiGraph,
    alpha: Sequence[Node],
    C: Iterable[Node],
) -> dict[str, object]:
    """
    非计时的一致性实验接口。

    ROS算法结束后，再在算法外构造最终图。
    """
    alpha_tuple = tuple(alpha)
    C_set = set(C)
    fill_edges = minimal_fill_edge_set(
        G=G,
        alpha=alpha_tuple,
        C_set=C_set,
    )
    return _build_ros_result_after_algorithm(
        G=G,
        C=C_set,
        E_alpha_R=fill_edges,
    )



def _timed_ros_once(
    G: nx.DiGraph,
    alpha: tuple[Node, ...],
    C_set: set[Node],
) -> tuple[list[Edge], int]:
    """
    单次ROS纯算法计时，单位为纳秒。

    计时在原始填充边列表生成后立即停止；G_R、最终图、
    排序去重、验证、比较和结果封装均不进入计时。
    """
    start = time.perf_counter_ns()
    fill_edges = minimal_fill_edge_set(
        G=G,
        alpha=alpha,
        C_set=C_set,
    )
    end = time.perf_counter_ns()
    return fill_edges, end - start


def _fcgr_added_edges_R(
    G: nx.DiGraph,
    C: Iterable[Node],
    H_R: nx.DiGraph,
) -> tuple[Edge, ...]:
    """
    在FCGR计时结束后计算R内部最终填充边集。

    该函数只供输出一致性实验和计时结果记录使用。
    """
    C_set = set(C)
    R_nodes = [
        node
        for node in G.nodes
        if node not in C_set
    ]
    G_R = G.subgraph(R_nodes)
    return tuple(
        sorted(
            set(H_R.edges()) - set(G_R.edges()),
            key=_edge_sort_key,
        )
    )
```

## 标准真实网络实验定义

```python
# ==========================================================
# 5. BIF 标准网络结构读取
# ==========================================================


EXPECTED_NODE_COUNTS = {
    "Asia": 8,
    "Insurance": 27,
    "Alarm": 37,
    "Win95pts": 76,
    "Andes": 223,
    "Munin": 1041,
}


def _clean_bif_name(raw_name: str) -> str:
    name = raw_name.strip()
    if (
        len(name) >= 2
        and name[0] == name[-1]
        and name[0] in {"'", '"'}
    ):
        name = name[1:-1]
    return name.strip()


def _load_bif_structure_fallback(path: Path) -> nx.DiGraph:
    """无 pgmpy 时，只读取 BIF 的变量名和父子关系。"""
    text = path.read_text(encoding="utf-8", errors="ignore")
    text = re.sub(r"/\*.*?\*/", "", text, flags=re.S)
    text = re.sub(r"//.*?$", "", text, flags=re.M)

    G = nx.DiGraph(name=path.stem)
    variable_pattern = re.compile(
        r"\bvariable\s+(\"[^\"]+\"|'[^']+'|[^\s\{]+)\s*\{",
        flags=re.I,
    )
    for match in variable_pattern.finditer(text):
        node = _clean_bif_name(match.group(1))
        if node:
            G.add_node(node)

    probability_pattern = re.compile(
        r"\bprobability\s*\(\s*(.*?)\s*\)\s*\{",
        flags=re.I | re.S,
    )
    for match in probability_pattern.finditer(text):
        header = match.group(1).strip()
        if "|" in header:
            child_text, parent_text = header.split("|", 1)
            parents = [
                _clean_bif_name(item)
                for item in parent_text.split(",")
                if _clean_bif_name(item)
            ]
        else:
            child_text = header
            parents = []

        child = _clean_bif_name(child_text)
        if not child:
            continue
        G.add_node(child)
        for parent in parents:
            G.add_edge(parent, child)

    if G.number_of_nodes() == 0:
        raise ValueError(f"未能从 BIF 文件读取任何节点：{path}")
    return G


def load_bif_dag(path: str | Path) -> nx.DiGraph:
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"BIF 文件不存在：{path}")

    try:
        from pgmpy.readwrite import BIFReader
    except ImportError:
        G = _load_bif_structure_fallback(path)
    else:
        model = BIFReader(str(path)).get_model()
        G = nx.DiGraph(name=path.stem)
        G.add_nodes_from(model.nodes())
        G.add_edges_from(model.edges())

    _validate_dag(G)
    return G


def load_standard_networks(
    data_dir: str | Path,
    network_files: Mapping[str, str],
    strict_node_count: bool = True,
) -> dict[str, nx.DiGraph]:
    data_dir = Path(data_dir)
    graphs: dict[str, nx.DiGraph] = {}

    for network, filename in network_files.items():
        G = load_bif_dag(data_dir / filename)
        G.name = network
        expected_n = EXPECTED_NODE_COUNTS.get(network)
        if (
            strict_node_count
            and expected_n is not None
            and G.number_of_nodes() != expected_n
        ):
            raise ValueError(
                f"{network} 节点数不正确："
                f"读取到 {G.number_of_nodes()}，预期 {expected_n}。"
            )
        graphs[network] = G

    return graphs


# ==========================================================
# 6. 实验 A：720 个边集与输出图一致性实例
# ==========================================================


@dataclass(frozen=True)
class StandardConsistencyConfig:
    """
    每个网络使用30个三节点条件集和4个固定拓扑序。

    实例总数：
        6个网络 * 30个条件集 * 4个拓扑序 = 720。
    """

    c_size: int = 3
    condition_sets_per_network: int = 30
    n_fixed_orders: int = 4
    condition_base_seed: int = 20260729
    order_base_seed: int = 20260812
    stop_on_mismatch: bool = False


def _validate_consistency_config(
    config: StandardConsistencyConfig,
) -> None:
    if config.c_size <= 0:
        raise ValueError("c_size 必须为正整数。")
    if config.condition_sets_per_network <= 0:
        raise ValueError("condition_sets_per_network 必须为正整数。")
    if config.n_fixed_orders <= 0:
        raise ValueError("n_fixed_orders 必须为正整数。")


def generate_fixed_random_topological_orders(
    G: nx.DiGraph,
    n_orders: int,
    seed_base: int,
    max_attempts: int = 10_000,
) -> list[dict[str, object]]:
    """生成可复现且互不相同的固定合法拓扑序。"""
    if n_orders <= 0:
        raise ValueError("n_orders 必须为正整数。")

    orders: list[dict[str, object]] = []
    seen: set[tuple[Node, ...]] = set()
    attempt = 0

    while len(orders) < n_orders and attempt < max_attempts:
        beta_seed = seed_base + attempt
        beta = tuple(random_topological_order(G=G, seed=beta_seed))
        if beta not in seen:
            seen.add(beta)
            orders.append(
                {
                    "order_id": len(orders) + 1,
                    "beta_seed": beta_seed,
                    "beta": beta,
                }
            )
        attempt += 1

    if len(orders) < n_orders:
        raise RuntimeError(
            f"只能生成 {len(orders)} 个不同拓扑序，"
            f"未达到要求的 {n_orders} 个。"
        )
    return orders


def _standard_mismatch_evidence(
    network: str,
    G: nx.DiGraph,
    condition_id: int,
    order_id: int,
    C_set: set[Node],
    beta: Sequence[Node],
    fcgr_fill: set[Edge],
    ros_fill: set[Edge],
    fcgr_output: set[Edge],
    ros_output: set[Edge],
) -> dict[str, object]:
    """
    构造标准网络不一致实例的完整证据。

    完整边集写入压缩JSONL，而不是写入Excel单元格，
    从而避免Munin等大网络超过Excel单元格长度上限。
    """
    # 将边集转换为稳定排序后的可JSON序列化格式。
    def edge_list(edges: Iterable[Edge]) -> list[list[str]]:
        return [
            [str(source), str(target)]
            for source, target in sorted(edges, key=_edge_sort_key)
        ]

    # 返回可独立复现该不一致实例的全部输入和输出。
    return {
        "network": network,
        "condition_id": condition_id,
        "order_id": order_id,
        "graph_fingerprint_sha256": _graph_fingerprint(G),
        "graph_edges": edge_list(set(G.edges())),
        "C": [str(node) for node in sorted(C_set, key=str)],
        "beta": [str(node) for node in beta],
        "FCGR_fill_edges": edge_list(fcgr_fill),
        "ROS_fill_edges": edge_list(ros_fill),
        "only_FCGR_fill_edges": edge_list(fcgr_fill - ros_fill),
        "only_ROS_fill_edges": edge_list(ros_fill - fcgr_fill),
        "FCGR_output_edges": edge_list(fcgr_output),
        "ROS_output_edges": edge_list(ros_output),
        "only_FCGR_output_edges": edge_list(fcgr_output - ros_output),
        "only_ROS_output_edges": edge_list(ros_output - fcgr_output),
    }


def run_one_consistency_case(
    network: str,
    G: nx.DiGraph,
    condition_id: int,
    order_id: int,
    C: Sequence[Node],
    beta: Sequence[Node],
    condition_seed: int,
    beta_seed: int,
) -> tuple[dict[str, object], dict[str, object] | None]:
    """
    在相同(G,C,beta)下运行FCGR与ROS并比较两个核心输出。

    核心判定：
        1. 有向填充边集完全一致；
        2. 最终输出图的节点集和有向边集完全一致。
    """
    # 固化条件集，保证两种算法使用同一输入。
    C_set = set(C)

    # 先写入可复现实验实例的元数据。
    row: dict[str, object] = {
        "network": network,
        "condition_id": condition_id,
        "order_id": order_id,
        "condition_seed": condition_seed,
        "beta_seed": beta_seed,
        "n_nodes": G.number_of_nodes(),
        "n_edges": G.number_of_edges(),
        "c_size": len(C_set),
        "C": _encode_nodes(C_set),
        "beta": _encode_order(beta),
        "status": "ok",
        "error_type": "",
        "error_message": "",
    }

    # 默认没有不一致证据；仅在失败时构造完整证据。
    mismatch_evidence: dict[str, object] | None = None

    try:
        # 运行不含验证的FCGR算法。
        fcgr_result = run_fcgr(
            G=G.copy(),
            C=C_set,
            beta=beta,
        )

        # 运行不含验证的ROS算法。
        ros_result = run_ros(
            G=G.copy(),
            alpha=beta,
            C=C_set,
        )

        # 提取两种算法的填充边集。
        fcgr_fill = set(
            _fcgr_added_edges_R(
                G=G,
                C=C_set,
                H_R=fcgr_result.H_R,
            )
        )
        ros_fill = set(ros_result["E_alpha_R"])

        # 提取两种算法的最终输出图。
        ros_graph = ros_result["G_alpha_R"]
        if not isinstance(ros_graph, nx.DiGraph):
            raise TypeError("ROS结果中的G_alpha_R必须是nx.DiGraph。")

        fcgr_nodes = set(fcgr_result.H_R.nodes())
        ros_nodes = set(ros_graph.nodes())
        fcgr_output = set(fcgr_result.H_R.edges())
        ros_output = set(ros_graph.edges())

        # 严格比较有向填充边集合。
        fill_equal = fcgr_fill == ros_fill

        # 严格比较最终节点集与完整有向边集合。
        graph_equal = (
            fcgr_nodes == ros_nodes
            and fcgr_output == ros_output
        )

        # Excel只保存数量和哈希，不保存大型完整边集。
        row.update(
            {
                "FCGR_fill_count": len(fcgr_fill),
                "ROS_fill_count": len(ros_fill),
                "FCGR_output_edge_count": len(fcgr_output),
                "ROS_output_edge_count": len(ros_output),
                "FCGR_fill_sha256": _hash_edges(fcgr_fill),
                "ROS_fill_sha256": _hash_edges(ros_fill),
                "FCGR_output_sha256": _hash_edges(fcgr_output),
                "ROS_output_sha256": _hash_edges(ros_output),
                "only_FCGR_fill_count": len(fcgr_fill - ros_fill),
                "only_ROS_fill_count": len(ros_fill - fcgr_fill),
                "only_FCGR_output_count": len(fcgr_output - ros_output),
                "only_ROS_output_count": len(ros_output - fcgr_output),
                "fill_equal": fill_equal,
                "graph_equal": graph_equal,
                "both_equal": fill_equal and graph_equal,
            }
        )

        # 只有出现差异时才保存完整边集证据。
        if not (fill_equal and graph_equal):
            mismatch_evidence = _standard_mismatch_evidence(
                network=network,
                G=G,
                condition_id=condition_id,
                order_id=order_id,
                C_set=C_set,
                beta=beta,
                fcgr_fill=fcgr_fill,
                ros_fill=ros_fill,
                fcgr_output=fcgr_output,
                ros_output=ros_output,
            )

    except Exception as exc:
        # 将运行异常视为一致性实验未通过。
        row.update(
            {
                "status": "error",
                "error_type": type(exc).__name__,
                "error_message": str(exc),
                "fill_equal": False,
                "graph_equal": False,
                "both_equal": False,
            }
        )

        # 异常证据至少保存完整输入，便于复现。
        mismatch_evidence = {
            "network": network,
            "condition_id": condition_id,
            "order_id": order_id,
            "graph_fingerprint_sha256": _graph_fingerprint(G),
            "C": [str(node) for node in sorted(C_set, key=str)],
            "beta": [str(node) for node in beta],
            "status": "error",
            "error_type": type(exc).__name__,
            "error_message": str(exc),
        }

    # 同时返回Excel行和可选的完整不一致证据。
    return row, mismatch_evidence


def _summarize_consistency(
    details: pd.DataFrame,
) -> pd.DataFrame:
    """
    汇总标准网络输出一致性结果。

    重要口径：
        所有实例都进入一致率分母；
        报错实例按False处理，不能从分母中排除。
    """
    summary_rows: list[dict[str, object]] = []

    # 按标准网络分别汇总。
    for network, group in details.groupby("network", sort=False):
        n_total = len(group)
        n_successful = int((group["status"] == "ok").sum())

        # 将缺失值和报错实例统一视为未通过。
        def rate_percent(column: str) -> float:
            values = group[column].fillna(False).astype(bool)
            return 100.0 * float(values.mean())

        # 任一报错、填充边不一致或最终图不一致均计为失败。
        mismatch_mask = (
            (group["status"] != "ok")
            | (~group["fill_equal"].fillna(False))
            | (~group["graph_equal"].fillna(False))
        )

        # 保存网络级汇总结果。
        summary_rows.append(
            {
                "network": network,
                "n_nodes": int(group["n_nodes"].iloc[0]),
                "n_edges": int(group["n_edges"].iloc[0]),
                "c_size": int(group["c_size"].iloc[0]),
                "n_condition_sets": group["condition_id"].nunique(),
                "n_fixed_orders": group["order_id"].nunique(),
                "n_instances": n_total,
                "n_successful": n_successful,
                "n_errors": n_total - n_successful,
                "fill_equal_rate_pct": rate_percent("fill_equal"),
                "graph_equal_rate_pct": rate_percent("graph_equal"),
                "both_equal_rate_pct": rate_percent("both_equal"),
                "n_mismatches": int(mismatch_mask.sum()),
            }
        )

    # 返回标准网络级结果表。
    return pd.DataFrame(summary_rows)


def run_standard_consistency_experiment(
    network_graphs: Mapping[str, nx.DiGraph],
    config: StandardConsistencyConfig = StandardConsistencyConfig(),
) -> tuple[
    pd.DataFrame,
    pd.DataFrame,
    pd.DataFrame,
    pd.DataFrame,
    list[dict[str, object]],
]:
    """
    完整运行标准网络输出一致性实验。

    每个网络的随机条件集与固定拓扑序完全交叉，
    主判定仅为填充边集一致和最终输出图一致。
    """
    # 检查实验配置。
    _validate_consistency_config(config)

    # 初始化实例结果、固定顺序和完整不一致证据。
    rows: list[dict[str, object]] = []
    order_rows: list[dict[str, object]] = []
    mismatch_evidence_rows: list[dict[str, object]] = []

    # 逐个标准网络执行实验。
    for network_index, (network, G) in enumerate(
        network_graphs.items(),
        start=1,
    ):
        # 在网络级别验证输入DAG。
        _validate_dag(G)

        # 检查条件集规模。
        if not 1 <= config.c_size < G.number_of_nodes():
            raise ValueError(
                f"{network}不满足1 <= c_size < |V|。"
            )

        # 分别派生条件集种子和拓扑序种子。
        condition_seed = (
            config.condition_base_seed
            + network_index * 1_000_000
            + 101
        )
        order_seed_base = (
            config.order_base_seed
            + network_index * 1_000_000
            + 10_000
        )

        # 随机抽取互不相同的条件集。
        condition_sets = sample_random_condition_sets(
            G=G,
            n_sets=config.condition_sets_per_network,
            c_size=config.c_size,
            seed=condition_seed,
        )
        if len(set(condition_sets)) != config.condition_sets_per_network:
            raise AssertionError(
                f"{network}未生成要求数量的互不相同条件集。"
            )

        # 为该网络生成固定随机拓扑序。
        fixed_orders = generate_fixed_random_topological_orders(
            G=G,
            n_orders=config.n_fixed_orders,
            seed_base=order_seed_base,
        )

        # 保存固定拓扑序，确保实验可复现。
        for item in fixed_orders:
            order_rows.append(
                {
                    "network": network,
                    "order_id": item["order_id"],
                    "beta_seed": item["beta_seed"],
                    "beta": _encode_order(item["beta"]),
                }
            )

        # 将每个条件集与每个固定拓扑序完全交叉。
        for condition_id, C in enumerate(condition_sets, start=1):
            for order_info in fixed_orders:
                row, evidence = run_one_consistency_case(
                    network=network,
                    G=G,
                    condition_id=condition_id,
                    order_id=int(order_info["order_id"]),
                    C=C,
                    beta=order_info["beta"],
                    condition_seed=condition_seed,
                    beta_seed=int(order_info["beta_seed"]),
                )
                rows.append(row)

                # 仅对失败实例保存完整证据。
                if evidence is not None:
                    mismatch_evidence_rows.append(evidence)

                # 可选的立即停止模式只依据两个核心输出。
                if config.stop_on_mismatch and not row["both_equal"]:
                    raise AssertionError(
                        f"{network}，条件集{condition_id}，"
                        f"拓扑序{order_info['order_id']}未通过一致性检查。"
                    )

        print(
            f"[一致性][{network}] 完成"
            f"{len(condition_sets) * len(fixed_orders)}个实例。"
        )

    # 构造实例级结果表与固定顺序表。
    details = pd.DataFrame(rows)
    fixed_orders_df = pd.DataFrame(order_rows)

    # 提取报错或输出不一致实例。
    mismatch_mask = (
        (details["status"] != "ok")
        | (~details["fill_equal"].fillna(False))
        | (~details["graph_equal"].fillna(False))
    )
    mismatches = details.loc[mismatch_mask].copy()

    # 汇总到网络级结果。
    summary = _summarize_consistency(details)

    # 检查实验实例数是否符合设计。
    expected_instances = (
        len(network_graphs)
        * config.condition_sets_per_network
        * config.n_fixed_orders
    )
    if len(details) != expected_instances:
        raise AssertionError(
            f"一致性实验实例数错误：得到{len(details)}，"
            f"预期{expected_instances}。"
        )

    # 返回表格结果和完整不一致证据。
    return (
        details,
        summary,
        mismatches,
        fixed_orders_df,
        mismatch_evidence_rows,
    )


def save_consistency_results(
    details: pd.DataFrame,
    summary: pd.DataFrame,
    mismatches: pd.DataFrame,
    fixed_orders: pd.DataFrame,
    mismatch_evidence: Sequence[Mapping[str, object]],
    config: StandardConsistencyConfig,
    output_file: str | Path,
    mismatch_file: str | Path,
) -> tuple[Path, Path]:
    """
    保存标准网络一致性实验结果。

    Excel保存汇总、哈希和计数；
    压缩JSONL保存完整不一致边集证据。
    """
    # 规范化输出路径并创建目录。
    output_file = Path(output_file)
    mismatch_file = Path(mismatch_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)
    mismatch_file.parent.mkdir(parents=True, exist_ok=True)

    # 将配置对象转换为可审计表格。
    config_df = pd.DataFrame(
        {
            "parameter": list(asdict(config).keys()),
            "value": list(asdict(config).values()),
        }
    )

    # 保存Excel主结果。
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(
            writer,
            sheet_name="network_summary",
            index=False,
        )
        fixed_orders.to_excel(
            writer,
            sheet_name="fixed_orders",
            index=False,
        )
        details.to_excel(
            writer,
            sheet_name="instance_results",
            index=False,
        )
        mismatches.to_excel(
            writer,
            sheet_name="mismatches",
            index=False,
        )
        config_df.to_excel(
            writer,
            sheet_name="config",
            index=False,
        )

    # 保存完整不一致证据；无失败时生成空压缩文件。
    with gzip.open(mismatch_file, mode="wt", encoding="utf-8") as handle:
        for evidence in mismatch_evidence:
            handle.write(
                json.dumps(
                    evidence,
                    ensure_ascii=False,
                    separators=(",", ":"),
                )
            )
            handle.write("\n")

    # 返回两个实际保存路径。
    return output_file, mismatch_file


# ==========================================================
# 7. 实验 B：180 个独立运行时间实例
# ==========================================================


@dataclass(frozen=True)
class RuntimeExperimentConfig:
    """标准网络无验证计时实验配置。"""

    # 六个网络统一使用三节点条件集。
    c_size: int = 3

    # 每个网络随机抽取30个互不相同的条件集。
    condition_sets_per_network: int = 30

    # 正式计时前执行5次预热，预热不写入正式结果。
    warmup_repetitions: int = 5

    # 每个(G,C,beta)内部重复30次并取中位数。
    timing_repetitions: int = 30

    # 控制每个网络的随机条件集。
    condition_base_seed: int = 20260729

    # 控制每个网络唯一固定的拓扑序beta。
    order_base_seed: int = 20260812



def _validate_runtime_config(config: RuntimeExperimentConfig) -> None:
    if config.c_size <= 0:
        raise ValueError("c_size 必须为正整数。")
    if config.condition_sets_per_network <= 0:
        raise ValueError("condition_sets_per_network 必须为正整数。")
    if config.warmup_repetitions < 0:
        raise ValueError("warmup_repetitions 不能为负数。")
    if config.timing_repetitions <= 0:
        raise ValueError("timing_repetitions 必须为正整数。")


def sample_random_condition_sets(
    G: nx.DiGraph,
    n_sets: int,
    c_size: int,
    seed: int,
) -> list[tuple[Node, ...]]:
    """
    固定随机种子抽取条件集。

    组合总数不少于 n_sets 时，无放回抽取，保证全部互不相同。
    组合总数少于 n_sets 时，先使用全部不同组合，再补充重复查询。

    本实验固定c_size=3。Asia有C(8,3)=56个候选条件集，
    因此六个网络均能抽取30个互不相同的条件集。
    """
    nodes = list(G.nodes())
    if not 1 <= c_size < len(nodes):
        raise ValueError("c_size 必须满足 1 <= c_size < |V|。")

    rng = random.Random(seed)
    max_unique = math.comb(len(nodes), c_size)
    unique_target = min(n_sets, max_unique)

    condition_sets: list[tuple[Node, ...]] = []
    seen: set[tuple[Node, ...]] = set()
    max_attempts = max(1000, 50 * unique_target)
    attempts = 0

    while len(condition_sets) < unique_target and attempts < max_attempts:
        attempts += 1
        C = tuple(sorted(rng.sample(nodes, c_size), key=str))
        if C not in seen:
            seen.add(C)
            condition_sets.append(C)

    if len(condition_sets) != unique_target:
        raise RuntimeError("无法生成要求数量的互不相同条件集。")

    unique_sets = list(condition_sets)
    while len(condition_sets) < n_sets:
        condition_sets.append(rng.choice(unique_sets))

    rng.shuffle(condition_sets)
    return condition_sets




def _run_pair_without_gc(
    G: nx.DiGraph,
    C_set: set[Node],
    beta: tuple[Node, ...],
    fcgr_first: bool,
) -> tuple[FCGRResult, list[Edge], int, int, int]:
    """
    在关闭自动垃圾回收的条件下运行一对无验证计时。

        返回：
        FCGR结果、ROS原始填充边列表、FCGR算法1时间、FCGR算法2时间、ROS总时间。
    """
    # 记录原垃圾回收状态。
    gc_was_enabled = gc.isenabled()

    # 计时期间关闭自动垃圾回收，降低随机暂停干扰。
    if gc_was_enabled:
        gc.disable()

    try:
        # 交替算法先后顺序，减弱缓存与固定先运行优势。
        if fcgr_first:
            fcgr_result, algorithm1_ns, algorithm2_ns = (
                _timed_fcgr_once(
                    G=G,
                    C_set=C_set,
                    beta=beta,
                )
            )
            ros_result, ros_total_ns = _timed_ros_once(
                G=G,
                alpha=beta,
                C_set=C_set,
            )
        else:
            ros_result, ros_total_ns = _timed_ros_once(
                G=G,
                alpha=beta,
                C_set=C_set,
            )
            fcgr_result, algorithm1_ns, algorithm2_ns = (
                _timed_fcgr_once(
                    G=G,
                    C_set=C_set,
                    beta=beta,
                )
            )
    finally:
        # 恢复调用前的垃圾回收状态。
        if gc_was_enabled:
            gc.enable()

    # 返回配对结果和原始纳秒计时。
    return (
        fcgr_result,
        ros_result,
        algorithm1_ns,
        algorithm2_ns,
        ros_total_ns,
    )



def _warm_up_network(
    G: nx.DiGraph,
    C: Sequence[Node],
    beta: Sequence[Node],
    repetitions: int,
) -> None:
    """正式计时前只执行算法预热，不做输入或输出验证。"""
    C_set = set(C)
    beta_tuple = tuple(beta)
    for warmup_id in range(1, repetitions + 1):
        gc.collect()
        _run_pair_without_gc(
            G=G,
            C_set=C_set,
            beta=beta_tuple,
            fcgr_first=warmup_id % 2 == 1,
        )




def run_one_runtime_case(
    network: str,
    G: nx.DiGraph,
    condition_id: int,
    C: Sequence[Node],
    beta: Sequence[Node],
    condition_seed: int,
    beta_seed: int,
    timing_repetitions: int,
) -> tuple[dict[str, object], list[dict[str, object]]]:
    """
    对固定(G,C,beta)执行纯算法配对计时。

    本函数不进行基准验证或重复后输出比较。FCGR最终填充边集只在
    全部计时结束后计算一次，因此不进入算法1、算法2或FCGR总时间。
    """
    C_set = set(C)
    beta_tuple = tuple(beta)
    algorithm1_times_ns: list[int] = []
    algorithm2_times_ns: list[int] = []
    fcgr_total_times_ns: list[int] = []
    ros_total_times_ns: list[int] = []
    repeat_rows: list[dict[str, object]] = []
    last_fcgr_result: FCGRResult | None = None
    last_ros_edges: list[Edge] | None = None

    for repeat_id in range(1, timing_repetitions + 1):
        gc.collect()
        fcgr_first = (condition_id + repeat_id) % 2 == 0
        (
            fcgr_result,
            ros_result,
            algorithm1_ns,
            algorithm2_ns,
            ros_total_ns,
        ) = _run_pair_without_gc(
            G=G,
            C_set=C_set,
            beta=beta_tuple,
            fcgr_first=fcgr_first,
        )
        fcgr_total_ns = algorithm1_ns + algorithm2_ns
        algorithm1_times_ns.append(algorithm1_ns)
        algorithm2_times_ns.append(algorithm2_ns)
        fcgr_total_times_ns.append(fcgr_total_ns)
        ros_total_times_ns.append(ros_total_ns)
        repeat_rows.append(
            {
                "network": network,
                "condition_id": condition_id,
                "repeat_id": repeat_id,
                "timing_scope": "pure_algorithm_only",
                "run_order": (
                    "FCGR_then_ROS"
                    if fcgr_first
                    else "ROS_then_FCGR"
                ),
                "FCGR_algorithm1_ms": algorithm1_ns / 1_000_000,
                "FCGR_algorithm2_ms": algorithm2_ns / 1_000_000,
                "FCGR_total_ms": fcgr_total_ns / 1_000_000,
                "ROS_total_ms": ros_total_ns / 1_000_000,
                "runtime_ratio_FCGR_to_ROS": (
                    fcgr_total_ns / ros_total_ns
                ),
            }
        )
        last_fcgr_result = fcgr_result
        last_ros_edges = ros_result

    if last_fcgr_result is None or last_ros_edges is None:
        raise RuntimeError("该条件集没有产生任何正式计时结果。")

    algorithm1_median_ns = float(median(algorithm1_times_ns))
    algorithm2_median_ns = float(median(algorithm2_times_ns))
    fcgr_total_median_ns = float(median(fcgr_total_times_ns))
    ros_total_median_ns = float(median(ros_total_times_ns))

    # 明确位于全部计时区间之外。
    fcgr_fill_count = len(
        _fcgr_added_edges_R(
            G=G,
            C=C_set,
            H_R=last_fcgr_result.H_R,
        )
    )
    ros_fill_count = len(last_ros_edges)

    condition_row: dict[str, object] = {
        "network": network,
        "condition_id": condition_id,
        "condition_seed": condition_seed,
        "beta_seed": beta_seed,
        "n_nodes": G.number_of_nodes(),
        "n_edges": G.number_of_edges(),
        "c_size": len(C_set),
        "C": _encode_nodes(C_set),
        "timing_scope": "pure_algorithm_only",
        "timing_repetitions": timing_repetitions,
        "FCGR_algorithm1_median_ms": algorithm1_median_ns / 1_000_000,
        "FCGR_algorithm2_median_ms": algorithm2_median_ns / 1_000_000,
        "FCGR_total_median_ms": fcgr_total_median_ns / 1_000_000,
        "ROS_total_median_ms": ros_total_median_ns / 1_000_000,
        "runtime_ratio_FCGR_to_ROS": (
            fcgr_total_median_ns / ros_total_median_ns
        ),
        "FCGR_fill_count_after_timing": fcgr_fill_count,
        "ROS_fill_count": ros_fill_count,
    }
    return condition_row, repeat_rows




def _summarize_runtime(
    details: pd.DataFrame,
) -> pd.DataFrame:
    """按网络汇总30个条件集实例的纯算法代表时间。"""
    summary_rows: list[dict[str, object]] = []
    for network, group in details.groupby("network", sort=False):
        fcgr_algorithm1_mean = float(
            group["FCGR_algorithm1_median_ms"].mean()
        )
        fcgr_algorithm2_mean = float(
            group["FCGR_algorithm2_median_ms"].mean()
        )
        fcgr_total_mean = float(
            group["FCGR_total_median_ms"].mean()
        )
        ros_total_mean = float(
            group["ROS_total_median_ms"].mean()
        )
        summary_rows.append(
            {
                "network": network,
                "n_nodes": int(group["n_nodes"].iloc[0]),
                "n_edges": int(group["n_edges"].iloc[0]),
                "c_size": int(group["c_size"].iloc[0]),
                "n_condition_sets": len(group),
                "n_unique_condition_sets": group["C"].nunique(),
                "timing_scope": "pure_algorithm_only",
                "FCGR_algorithm1_mean_ms": fcgr_algorithm1_mean,
                "FCGR_algorithm2_mean_ms": fcgr_algorithm2_mean,
                "FCGR_total_mean_ms": fcgr_total_mean,
                "ROS_total_mean_ms": ros_total_mean,
                "runtime_ratio_FCGR_to_ROS": (
                    fcgr_total_mean / ros_total_mean
                ),
            }
        )
    return pd.DataFrame(summary_rows)




def run_runtime_experiment(
    network_graphs: Mapping[str, nx.DiGraph],
    config: RuntimeExperimentConfig = RuntimeExperimentConfig(),
) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """运行六个标准网络的纯算法配对时间实验。"""
    _validate_runtime_config(config)
    condition_rows: list[dict[str, object]] = []
    repeat_rows: list[dict[str, object]] = []
    fixed_input_rows: list[dict[str, object]] = []

    for network_index, (network, G) in enumerate(
        network_graphs.items(),
        start=1,
    ):
        # 这里只验证实验输入，不属于FCGR或ROS，也不进入计时。
        _validate_dag(G)
        if not 1 <= config.c_size < G.number_of_nodes():
            raise ValueError(
                f"{network} 不满足 1 <= c_size < |V|。"
            )

        condition_seed = (
            config.condition_base_seed
            + network_index * 1_000_000
            + 101
        )
        beta_seed = (
            config.order_base_seed
            + network_index * 1_000_000
            + 10_000
        )
        beta = tuple(random_topological_order(G, seed=beta_seed))
        condition_sets = sample_random_condition_sets(
            G=G,
            n_sets=config.condition_sets_per_network,
            c_size=config.c_size,
            seed=condition_seed,
        )
        if len(set(condition_sets)) != config.condition_sets_per_network:
            raise AssertionError(
                f"{network}未生成30个互不相同的条件集。"
            )

        fixed_input_rows.append(
            {
                "network": network,
                "n_nodes": G.number_of_nodes(),
                "n_edges": G.number_of_edges(),
                "condition_seed": condition_seed,
                "beta_seed": beta_seed,
                "c_size": config.c_size,
                "n_condition_sets": len(condition_sets),
                "n_unique_condition_sets": len(set(condition_sets)),
                "beta": _encode_order(beta),
            }
        )

        _warm_up_network(
            G=G,
            C=condition_sets[0],
            beta=beta,
            repetitions=config.warmup_repetitions,
        )

        for condition_id, C in enumerate(condition_sets, start=1):
            condition_row, current_repeat_rows = run_one_runtime_case(
                network=network,
                G=G,
                condition_id=condition_id,
                C=C,
                beta=beta,
                condition_seed=condition_seed,
                beta_seed=beta_seed,
                timing_repetitions=config.timing_repetitions,
            )
            condition_rows.append(condition_row)
            repeat_rows.extend(current_repeat_rows)

        current_details = pd.DataFrame(
            row
            for row in condition_rows
            if row["network"] == network
        )
        current_summary = _summarize_runtime(current_details).iloc[0]
        print(
            f"[计时][{network}] 完成{len(condition_sets)}个条件集实例；"
            f"FCGR={current_summary['FCGR_total_mean_ms']:.6f} ms，"
            f"ROS={current_summary['ROS_total_mean_ms']:.6f} ms。"
        )

    details = pd.DataFrame(condition_rows)
    repeat_times = pd.DataFrame(repeat_rows)
    fixed_inputs = pd.DataFrame(fixed_input_rows)
    summary = _summarize_runtime(details)
    expected_instances = (
        len(network_graphs)
        * config.condition_sets_per_network
    )
    if len(details) != expected_instances:
        raise AssertionError(
            f"计时实验实例数错误：得到{len(details)}，"
            f"预期{expected_instances}。"
        )
    return summary, details, repeat_times, fixed_inputs



def save_runtime_results(
    summary: pd.DataFrame,
    details: pd.DataFrame,
    repeat_times: pd.DataFrame,
    fixed_inputs: pd.DataFrame,
    config: RuntimeExperimentConfig,
    output_file: str | Path,
) -> Path:
    """将主表、条件集级结果、原始计时、固定输入和配置保存到 Excel。"""
    output_file = Path(output_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)

    config_df = pd.DataFrame(
        {
            "parameter": list(asdict(config).keys()),
            "value": list(asdict(config).values()),
        }
    )

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(
            writer,
            sheet_name="network_summary",
            index=False,
        )
        details.to_excel(
            writer,
            sheet_name="condition_details",
            index=False,
        )
        repeat_times.to_excel(
            writer,
            sheet_name="raw_repeat_times",
            index=False,
        )
        fixed_inputs.to_excel(
            writer,
            sheet_name="fixed_inputs",
            index=False,
        )
        config_df.to_excel(
            writer,
            sheet_name="config",
            index=False,
        )

    return output_file


```

```python
# 标准网络实验配置：720个一致性实例与180个计时实例彼此独立。
STANDARD_DATA_DIR = Path(
    r"D:\jupyternoteboook\条件化实验\ROS and\data"
)

STANDARD_NETWORK_FILES = {
    "Asia": "asia.bif",
    "Insurance": "insurance.bif",
    "Alarm": "alarm.bif",
    "Win95pts": "win95pts.bif",
    "Andes": "andes.bif",
    "Munin": "munin.bif",
}

STANDARD_CONSISTENCY_CONFIG = StandardConsistencyConfig(
    c_size=3,
    condition_sets_per_network=30,
    n_fixed_orders=4,
    condition_base_seed=20260729,
    order_base_seed=20260812,
    stop_on_mismatch=False,
)

STANDARD_RUNTIME_CONFIG = RuntimeExperimentConfig(
    c_size=3,
    condition_sets_per_network=30,
    warmup_repetitions=5,
    timing_repetitions=30,
    condition_base_seed=20260729,
    order_base_seed=20260812,
)

STANDARD_RESULT_DIR = Path("./results_standard_networks")
STANDARD_CONSISTENCY_RESULT_FILE = (
    STANDARD_RESULT_DIR / "FCGR_ROS_consistency_720_cases.xlsx"
)

# 完整不一致边集单独写入压缩JSONL，避免Excel单元格过长。
STANDARD_CONSISTENCY_MISMATCH_FILE = (
    STANDARD_RESULT_DIR / "FCGR_ROS_consistency_mismatches.jsonl.gz"
)
STANDARD_RUNTIME_RESULT_FILE = (
    STANDARD_RESULT_DIR / "FCGR_ROS_runtime_180_cases.xlsx"
)


def run_standard_consistency_only(
    data_dir: str | Path = STANDARD_DATA_DIR,
    network_files: Mapping[str, str] = STANDARD_NETWORK_FILES,
    config: StandardConsistencyConfig = STANDARD_CONSISTENCY_CONFIG,
    output_file: str | Path = STANDARD_CONSISTENCY_RESULT_FILE,
    mismatch_file: str | Path = STANDARD_CONSISTENCY_MISMATCH_FILE,
) -> dict[str, object]:
    """
    独立运行标准网络输出一致性实验。

    Excel保存计数和哈希；
    完整失败证据保存到压缩JSONL。
    """
    # 读取并验证六个标准网络。
    network_graphs = load_standard_networks(
        data_dir=data_dir,
        network_files=network_files,
        strict_node_count=True,
    )

    # 运行全部条件集与固定拓扑序的交叉实验。
    (
        details,
        summary,
        mismatches,
        fixed_orders,
        mismatch_evidence,
    ) = run_standard_consistency_experiment(
        network_graphs=network_graphs,
        config=config,
    )

    # 保存Excel结果和完整不一致证据。
    saved_path, saved_mismatch_path = save_consistency_results(
        details=details,
        summary=summary,
        mismatches=mismatches,
        fixed_orders=fixed_orders,
        mismatch_evidence=mismatch_evidence,
        config=config,
        output_file=output_file,
        mismatch_file=mismatch_file,
    )

    # 在Notebook中显示关键结果路径。
    print(summary.to_string(index=False))
    print(f"不一致或报错实例数：{len(mismatches)}")
    print(f"Excel结果：{saved_path.resolve()}")
    print(f"完整失败证据：{saved_mismatch_path.resolve()}")

    # 返回全部内存结果，便于后续分析。
    return {
        "summary": summary,
        "details": details,
        "mismatches": mismatches,
        "fixed_orders": fixed_orders,
        "mismatch_evidence": mismatch_evidence,
        "output_file": saved_path,
        "mismatch_file": saved_mismatch_path,
    }


def run_standard_runtime_only(
    data_dir: str | Path = STANDARD_DATA_DIR,
    network_files: Mapping[str, str] = STANDARD_NETWORK_FILES,
    config: RuntimeExperimentConfig = STANDARD_RUNTIME_CONFIG,
    output_file: str | Path = STANDARD_RUNTIME_RESULT_FILE,
) -> dict[str, object]:
    """
    独立运行标准网络无验证计时实验。

    输入验证、结果审计和输出一致性比较均在计时区间之外完成。
    """
    # 读取并验证六个标准网络。
    network_graphs = load_standard_networks(
        data_dir=data_dir,
        network_files=network_files,
        strict_node_count=True,
    )

    # 运行标准网络无验证配对计时。
    summary, details, repeat_times, fixed_inputs = (
        run_runtime_experiment(
            network_graphs=network_graphs,
            config=config,
        )
    )

    # 保存汇总、实例级结果和原始重复时间。
    saved_path = save_runtime_results(
        summary=summary,
        details=details,
        repeat_times=repeat_times,
        fixed_inputs=fixed_inputs,
        config=config,
        output_file=output_file,
    )

    # 显示主结果和文件位置。
    print(summary.to_string(index=False))
    print(f"结果文件：{saved_path.resolve()}")

    # 返回全部结果对象。
    return {
        "summary": summary,
        "details": details,
        "repeat_times": repeat_times,
        "fixed_inputs": fixed_inputs,
        "output_file": saved_path,
    }
```

## 大量随机网络实验定义

```python
# ==========================================================
# 5. 随机 DAG 设计与固定输入生成
# ==========================================================


@dataclass(frozen=True)
class RandomConsistencyConfig:
    node_sizes: tuple[int, ...] = (20, 50, 100, 200, 500, 1000)
    target_mean_indegrees: tuple[float, ...] = (1.0, 2.0, 4.0)
    graphs_per_cell: int = 30
    c_size: int = 3
    n_fixed_orders: int = 4
    root_seed: int = 20260730
    topological_order_attempts: int = 10_000
    graph_retry_limit: int = 100
    stop_on_mismatch: bool = False


@dataclass(frozen=True)
class RandomRuntimeConfig:
    """随机网络无验证计时实验配置。"""

    # 随机DAG节点规模。
    node_sizes: tuple[int, ...] = (20, 50, 100, 200, 500, 1000)

    # 随机DAG目标平均入度。
    target_mean_indegrees: tuple[float, ...] = (1.0, 2.0, 4.0)

    # 每个(n,d)组合生成30张独立随机图。
    graphs_per_cell: int = 30

    # 每张随机图随机抽取3个条件节点。
    c_size: int = 3

    # 每张图生成4个固定随机拓扑序；计时使用第1个。
    n_fixed_orders: int = 4

    # 控制全部随机图、条件集和拓扑序的根种子。
    root_seed: int = 20260730

    # 生成不同拓扑序时允许的最大尝试次数。
    topological_order_attempts: int = 10_000

    # 随机图生成失败时允许的最大重试次数。
    graph_retry_limit: int = 100

    # 每个(n,d)组合正式计时前执行5次预热。
    warmup_repetitions: int = 5

    # 每张随机图内部重复30次无验证计时。
    timing_repetitions: int = 30

    # 配对bootstrap重复次数。
    bootstrap_repetitions: int = 10_000



@dataclass(frozen=True)
class FixedOrder:
    order_id: int
    beta: tuple[Node, ...]
    beta_seed: int | None
    generation_method: str


@dataclass
class RandomGraphCase:
    n_nodes: int
    target_mean_indegree: float
    graph_id: int
    generation_attempt: int
    graph_seed: int
    condition_seed: int
    order_seed_base: int
    G: nx.DiGraph
    C: tuple[Node, ...]
    orders: tuple[FixedOrder, ...]


def _validate_design_config(
    config: RandomConsistencyConfig | RandomRuntimeConfig,
) -> None:
    if not config.node_sizes:
        raise ValueError("node_sizes 不能为空。")
    if not config.target_mean_indegrees:
        raise ValueError("target_mean_indegrees 不能为空。")
    if config.graphs_per_cell <= 0:
        raise ValueError("graphs_per_cell 必须为正整数。")
    if config.c_size <= 0:
        raise ValueError("c_size 必须为正整数。")
    if config.n_fixed_orders <= 0:
        raise ValueError("n_fixed_orders 必须为正整数。")
    if config.topological_order_attempts <= 0:
        raise ValueError("topological_order_attempts 必须为正整数。")
    if config.graph_retry_limit <= 0:
        raise ValueError("graph_retry_limit 必须为正整数。")

    for n_nodes in config.node_sizes:
        if n_nodes <= config.c_size:
            raise ValueError(
                f"n={n_nodes} 必须大于 c_size={config.c_size}。"
            )
        for target_d in config.target_mean_indegrees:
            edge_probability = 2.0 * target_d / (n_nodes - 1)
            if not 0.0 <= edge_probability <= 1.0:
                raise ValueError(
                    f"(n={n_nodes}, d={target_d}) 对应的加边概率"
                    f" p={edge_probability} 不在 [0,1]。"
                )

    if isinstance(config, RandomRuntimeConfig):
        if config.warmup_repetitions < 0:
            raise ValueError("warmup_repetitions 不能为负数。")
        if config.timing_repetitions <= 0:
            raise ValueError("timing_repetitions 必须为正整数。")
        if config.bootstrap_repetitions <= 0:
            raise ValueError("bootstrap_repetitions 必须为正整数。")


def _design_signature(
    config: RandomConsistencyConfig | RandomRuntimeConfig,
) -> tuple[object, ...]:
    """判断两项实验是否会重建同一批随机输入。"""
    return (
        config.node_sizes,
        config.target_mean_indegrees,
        config.graphs_per_cell,
        config.c_size,
        config.n_fixed_orders,
        config.root_seed,
        config.topological_order_attempts,
        config.graph_retry_limit,
    )


@lru_cache(maxsize=None)
def _upper_triangle_indices(
    n_nodes: int,
) -> tuple[np.ndarray, np.ndarray]:
    """缓存 i<j 的全部位置，避免每张图重复构造索引。"""
    return np.triu_indices(n_nodes, k=1)


def generate_random_dag(
    n_nodes: int,
    target_mean_indegree: float,
    seed: int,
) -> nx.DiGraph:
    """
    生成 G(n,p) 型随机 DAG。

    先随机排列节点；只允许排列中较早的节点指向较晚的节点；
    每个允许方向以
        p = 2 * target_mean_indegree / (n_nodes - 1)
    独立加边，因此 E[|E|] = n_nodes * target_mean_indegree。
    """
    if n_nodes < 2:
        raise ValueError("n_nodes 必须至少为2。")

    edge_probability = (
        2.0 * float(target_mean_indegree) / (n_nodes - 1)
    )
    if not 0.0 <= edge_probability <= 1.0:
        raise ValueError("随机DAG的加边概率不在[0,1]。")

    rng = np.random.default_rng(seed)
    width = max(2, len(str(n_nodes)))
    nodes = np.asarray(
        [f"X{index:0{width}d}" for index in range(1, n_nodes + 1)],
        dtype=object,
    )
    latent_order = rng.permutation(nodes)
    source_positions, target_positions = _upper_triangle_indices(n_nodes)
    selected = rng.random(len(source_positions)) < edge_probability

    selected_sources = latent_order[source_positions[selected]]
    selected_targets = latent_order[target_positions[selected]]

    G = nx.DiGraph()
    G.add_nodes_from(nodes.tolist())
    G.add_edges_from(
        zip(
            selected_sources.tolist(),
            selected_targets.tolist(),
        )
    )
    G.graph["generation_order"] = tuple(latent_order.tolist())
    G.graph["edge_probability"] = edge_probability
    G.graph["target_mean_indegree"] = target_mean_indegree

    _validate_dag(G)
    return G


def generate_fixed_topological_orders(
    G: nx.DiGraph,
    n_orders: int,
    seed_base: int,
    max_attempts: int,
) -> tuple[FixedOrder, ...]:
    """生成互不相同的固定拓扑序；不足时用穷举器补齐前4个。"""
    seen: set[tuple[Node, ...]] = set()
    orders: list[FixedOrder] = []

    for attempt in range(max_attempts):
        beta_seed = _derive_seed(seed_base, "beta", attempt)
        beta = random_topological_order(G=G, seed=beta_seed)
        if beta not in seen:
            seen.add(beta)
            orders.append(
                FixedOrder(
                    order_id=len(orders) + 1,
                    beta=beta,
                    beta_seed=beta_seed,
                    generation_method="randomized_kahn",
                )
            )
            if len(orders) == n_orders:
                return tuple(orders)

    # 极少数情况下随机抽样未找到足够不同顺序，枚举前几个拓扑序补齐。
    for beta_list in nx.all_topological_sorts(G):
        beta = tuple(beta_list)
        if beta not in seen:
            seen.add(beta)
            orders.append(
                FixedOrder(
                    order_id=len(orders) + 1,
                    beta=beta,
                    beta_seed=None,
                    generation_method="enumerated_fallback",
                )
            )
            if len(orders) == n_orders:
                return tuple(orders)

    raise RuntimeError(
        f"该随机图只有 {len(orders)} 个可获得的不同拓扑序，"
        f"少于要求的 {n_orders} 个。"
    )


def build_random_graph_case(
    n_nodes: int,
    target_mean_indegree: float,
    graph_id: int,
    config: RandomConsistencyConfig | RandomRuntimeConfig,
) -> RandomGraphCase:
    """
    构造一张图、一个条件集和4个固定拓扑序。

    若某张图实际支持的不同拓扑序少于4个，则透明地重新抽图，
    并记录 generation_attempt 与最终 graph_seed。
    """
    for generation_attempt in range(config.graph_retry_limit):
        graph_seed = _derive_seed(
            config.root_seed,
            "graph",
            n_nodes,
            target_mean_indegree,
            graph_id,
            generation_attempt,
        )
        condition_seed = _derive_seed(
            config.root_seed,
            "condition",
            n_nodes,
            target_mean_indegree,
            graph_id,
            generation_attempt,
        )
        order_seed_base = _derive_seed(
            config.root_seed,
            "orders",
            n_nodes,
            target_mean_indegree,
            graph_id,
            generation_attempt,
        )

        G = generate_random_dag(
            n_nodes=n_nodes,
            target_mean_indegree=target_mean_indegree,
            seed=graph_seed,
        )
        G.name = (
            f"Random_n{n_nodes}_d{target_mean_indegree:g}"
            f"_g{graph_id:02d}"
        )

        rng = random.Random(condition_seed)
        C = tuple(
            sorted(
                rng.sample(list(G.nodes), config.c_size),
                key=str,
            )
        )

        try:
            orders = generate_fixed_topological_orders(
                G=G,
                n_orders=config.n_fixed_orders,
                seed_base=order_seed_base,
                max_attempts=config.topological_order_attempts,
            )
        except RuntimeError:
            continue

        return RandomGraphCase(
            n_nodes=n_nodes,
            target_mean_indegree=target_mean_indegree,
            graph_id=graph_id,
            generation_attempt=generation_attempt,
            graph_seed=graph_seed,
            condition_seed=condition_seed,
            order_seed_base=order_seed_base,
            G=G,
            C=C,
            orders=orders,
        )

    raise RuntimeError(
        f"(n={n_nodes}, d={target_mean_indegree}, graph_id={graph_id})"
        f" 在 {config.graph_retry_limit} 次内未生成支持"
        f" {config.n_fixed_orders} 个不同拓扑序的随机DAG。"
    )


def iter_random_graph_cases(
    config: RandomConsistencyConfig | RandomRuntimeConfig,
) -> Iterator[RandomGraphCase]:
    """按 (n,d,graph_id) 的固定顺序逐张重建随机图。"""
    _validate_design_config(config)
    for n_nodes in config.node_sizes:
        for target_d in config.target_mean_indegrees:
            for graph_id in range(1, config.graphs_per_cell + 1):
                yield build_random_graph_case(
                    n_nodes=n_nodes,
                    target_mean_indegree=target_d,
                    graph_id=graph_id,
                    config=config,
                )


def _case_manifest_row(case: RandomGraphCase) -> dict[str, object]:
    G = case.G
    undirected = G.to_undirected(as_view=True)
    return {
        "n_nodes": case.n_nodes,
        "target_mean_indegree": case.target_mean_indegree,
        "graph_id": case.graph_id,
        "generation_attempt": case.generation_attempt,
        "graph_seed": case.graph_seed,
        "condition_seed": case.condition_seed,
        "order_seed_base": case.order_seed_base,
        "n_edges": G.number_of_edges(),
        "actual_mean_indegree": G.number_of_edges() / G.number_of_nodes(),
        "edge_probability": G.graph["edge_probability"],
        "weak_component_count": nx.number_connected_components(undirected),
        "isolated_node_count": len(list(nx.isolates(G))),
        "graph_fingerprint_sha256": _graph_fingerprint(G),
        "c_size": len(case.C),
        "C": _encode_nodes(case.C),
        "C_sha256": _hash_nodes(case.C),
    }


def _fixed_order_rows(
    case: RandomGraphCase,
) -> list[dict[str, object]]:
    return [
        {
            "n_nodes": case.n_nodes,
            "target_mean_indegree": case.target_mean_indegree,
            "graph_id": case.graph_id,
            "order_id": order.order_id,
            "beta_seed": order.beta_seed,
            "generation_method": order.generation_method,
            "beta_sha256": _hash_order(order.beta),
            "beta": _encode_order(order.beta),
        }
        for order in case.orders
    ]


# ==========================================================
# 6. 实验 A：2160 个输出一致性实例
# ==========================================================



def _compare_outputs(
    G: nx.DiGraph,
    C_set: set[Node],
    fcgr_result: FCGRResult,
    ros_result: Mapping[str, object],
) -> dict[str, object]:
    """一致性实验的外部比较；不属于两个算法，也不进入计时。"""
    ros_graph = ros_result["G_alpha_R"]
    if not isinstance(ros_graph, nx.DiGraph):
        raise TypeError("ROS结果中的G_alpha_R必须是nx.DiGraph。")

    fcgr_fill = set(
        _fcgr_added_edges_R(
            G=G,
            C=C_set,
            H_R=fcgr_result.H_R,
        )
    )
    ros_fill = set(ros_result["E_alpha_R"])
    fcgr_nodes = set(fcgr_result.H_R.nodes())
    ros_nodes = set(ros_graph.nodes())
    fcgr_edges = set(fcgr_result.H_R.edges())
    ros_edges = set(ros_graph.edges())
    alpha_R_position = {
        node: index
        for index, node in enumerate(fcgr_result.orders.alpha_R)
    }
    fcgr_order_valid = all(
        alpha_R_position[source] < alpha_R_position[target]
        for source, target in fcgr_edges
    )
    ros_order_valid = all(
        alpha_R_position[source] < alpha_R_position[target]
        for source, target in ros_edges
    )
    return {
        "fill_equal": fcgr_fill == ros_fill,
        "nodes_equal": fcgr_nodes == ros_nodes,
        "graph_equal": (
            fcgr_nodes == ros_nodes and fcgr_edges == ros_edges
        ),
        "FCGR_is_dag": nx.is_directed_acyclic_graph(fcgr_result.H_R),
        "ROS_is_dag": nx.is_directed_acyclic_graph(ros_graph),
        "FCGR_order_valid": fcgr_order_valid,
        "ROS_order_valid": ros_order_valid,
        "FCGR_fill_count": len(fcgr_fill),
        "ROS_fill_count": len(ros_fill),
        "FCGR_output_edge_count": len(fcgr_edges),
        "ROS_output_edge_count": len(ros_edges),
        "FCGR_fill_sha256": _hash_edges(fcgr_fill),
        "ROS_fill_sha256": _hash_edges(ros_fill),
        "FCGR_output_sha256": _hash_edges(fcgr_edges),
        "ROS_output_sha256": _hash_edges(ros_edges),
        "_FCGR_fill_edges": fcgr_fill,
        "_ROS_fill_edges": ros_fill,
        "_FCGR_output_edges": fcgr_edges,
        "_ROS_output_edges": ros_edges,
    }



def _mismatch_evidence(
    case: RandomGraphCase,
    order: FixedOrder,
    comparison: Mapping[str, object],
) -> dict[str, object]:
    fcgr_fill = set(comparison["_FCGR_fill_edges"])
    ros_fill = set(comparison["_ROS_fill_edges"])
    fcgr_output = set(comparison["_FCGR_output_edges"])
    ros_output = set(comparison["_ROS_output_edges"])

    return {
        "n_nodes": case.n_nodes,
        "target_mean_indegree": case.target_mean_indegree,
        "graph_id": case.graph_id,
        "generation_attempt": case.generation_attempt,
        "graph_seed": case.graph_seed,
        "condition_seed": case.condition_seed,
        "order_seed_base": case.order_seed_base,
        "order_id": order.order_id,
        "beta_seed": order.beta_seed,
        "graph_fingerprint_sha256": _graph_fingerprint(case.G),
        "graph_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                case.G.edges,
                key=_edge_sort_key,
            )
        ],
        "C": [str(node) for node in case.C],
        "beta": [str(node) for node in order.beta],
        "FCGR_fill_edges": [
            [str(source), str(target)]
            for source, target in sorted(fcgr_fill, key=_edge_sort_key)
        ],
        "ROS_fill_edges": [
            [str(source), str(target)]
            for source, target in sorted(ros_fill, key=_edge_sort_key)
        ],
        "only_FCGR_fill_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                fcgr_fill - ros_fill,
                key=_edge_sort_key,
            )
        ],
        "only_ROS_fill_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                ros_fill - fcgr_fill,
                key=_edge_sort_key,
            )
        ],
        "FCGR_output_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                fcgr_output,
                key=_edge_sort_key,
            )
        ],
        "ROS_output_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                ros_output,
                key=_edge_sort_key,
            )
        ],
        "only_FCGR_output_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                fcgr_output - ros_output,
                key=_edge_sort_key,
            )
        ],
        "only_ROS_output_edges": [
            [str(source), str(target)]
            for source, target in sorted(
                ros_output - fcgr_output,
                key=_edge_sort_key,
            )
        ],
    }


def run_random_consistency_experiment(
    config: RandomConsistencyConfig = RandomConsistencyConfig(),
) -> tuple[
    pd.DataFrame,
    pd.DataFrame,
    pd.DataFrame,
    pd.DataFrame,
    list[dict[str, object]],
]:
    """
    独立运行随机网络输出一致性实验。

    主通过标准仅包含：
        1. 填充边集一致；
        2. 最终输出图一致。

    DAG与顺序检查仅作为附加诊断，不改变主一致性判定。
    """
    # 验证随机网络实验配置。
    _validate_design_config(config)

    # 初始化实例结果、输入清单、固定顺序和失败证据。
    instance_rows: list[dict[str, object]] = []
    manifest_rows: list[dict[str, object]] = []
    order_rows: list[dict[str, object]] = []
    mismatch_evidence_rows: list[dict[str, object]] = []

    # 逐张生成并处理随机DAG。
    for case in iter_random_graph_cases(config):
        # 保存随机图输入清单和四个固定拓扑序。
        manifest_rows.append(_case_manifest_row(case))
        order_rows.extend(_fixed_order_rows(case))

        # 每张图分别使用四个固定拓扑序。
        for order in case.orders:
            # 先记录可复现输入元数据。
            row: dict[str, object] = {
                "n_nodes": case.n_nodes,
                "target_mean_indegree": case.target_mean_indegree,
                "graph_id": case.graph_id,
                "order_id": order.order_id,
                "graph_seed": case.graph_seed,
                "condition_seed": case.condition_seed,
                "beta_seed": order.beta_seed,
                "graph_fingerprint_sha256": _graph_fingerprint(case.G),
                "C_sha256": _hash_nodes(case.C),
                "beta_sha256": _hash_order(order.beta),
                "n_edges": case.G.number_of_edges(),
                "c_size": len(case.C),
                "status": "ok",
                "error_type": "",
                "error_message": "",
            }

            try:
                # 运行不含验证的FCGR算法。
                fcgr_result = run_fcgr(
                    G=case.G,
                    C=case.C,
                    beta=order.beta,
                )

                # 运行不含验证的ROS算法。
                ros_result = run_ros(
                    G=case.G,
                    alpha=order.beta,
                    C=case.C,
                )

                # 比较核心输出并计算附加结构诊断。
                comparison = _compare_outputs(
                    G=case.G,
                    C_set=set(case.C),
                    fcgr_result=fcgr_result,
                    ros_result=ros_result,
                )

                # 主一致性判定仅依赖填充边和最终输出图。
                both_equal = bool(
                    comparison["fill_equal"]
                    and comparison["graph_equal"]
                )

                # 结构合法性只作为附加诊断字段。
                structurally_valid = bool(
                    comparison["FCGR_is_dag"]
                    and comparison["ROS_is_dag"]
                    and comparison["FCGR_order_valid"]
                    and comparison["ROS_order_valid"]
                )

                # 将公开比较字段写入结果表。
                row.update(
                    {
                        key: value
                        for key, value in comparison.items()
                        if not key.startswith("_")
                    }
                )
                row["both_equal"] = both_equal
                row["output_consistency_passed"] = both_equal
                row["structurally_valid"] = structurally_valid
                row["diagnostic_all_checks_passed"] = (
                    both_equal and structurally_valid
                )

                # 只有核心输出不一致时才写入完整边集证据。
                if not both_equal:
                    mismatch_evidence_rows.append(
                        _mismatch_evidence(
                            case=case,
                            order=order,
                            comparison=comparison,
                        )
                    )

            except Exception as exc:
                # 报错实例按输出一致性未通过处理。
                row.update(
                    {
                        "status": "error",
                        "error_type": type(exc).__name__,
                        "error_message": str(exc),
                        "fill_equal": False,
                        "nodes_equal": False,
                        "graph_equal": False,
                        "both_equal": False,
                        "output_consistency_passed": False,
                        "structurally_valid": False,
                        "diagnostic_all_checks_passed": False,
                    }
                )

                # 保存足以复现异常的输入信息。
                mismatch_evidence_rows.append(
                    {
                        "n_nodes": case.n_nodes,
                        "target_mean_indegree": (
                            case.target_mean_indegree
                        ),
                        "graph_id": case.graph_id,
                        "generation_attempt": case.generation_attempt,
                        "graph_seed": case.graph_seed,
                        "condition_seed": case.condition_seed,
                        "order_seed_base": case.order_seed_base,
                        "order_id": order.order_id,
                        "beta_seed": order.beta_seed,
                        "graph_fingerprint_sha256": (
                            _graph_fingerprint(case.G)
                        ),
                        "C": [str(node) for node in case.C],
                        "beta": [str(node) for node in order.beta],
                        "status": "error",
                        "error_type": type(exc).__name__,
                        "error_message": str(exc),
                    }
                )

            # 保存当前配对实例结果。
            instance_rows.append(row)

            # 立即停止只依据两个核心输出。
            if (
                config.stop_on_mismatch
                and not row["output_consistency_passed"]
            ):
                raise AssertionError(
                    "随机网络输出一致性检查失败："
                    f"n={case.n_nodes}, "
                    f"d={case.target_mean_indegree}, "
                    f"graph_id={case.graph_id}, "
                    f"order_id={order.order_id}, "
                    f"error={row['error_message']}"
                )

        # 每个(n,d)单元完成后显示进度。
        if case.graph_id == config.graphs_per_cell:
            print(
                "[一致性] "
                f"n={case.n_nodes}, "
                f"d={case.target_mean_indegree:g}："
                f"{config.graphs_per_cell * config.n_fixed_orders}"
                "个实例完成。"
            )

    # 构造实例结果、图清单和固定顺序表。
    details = pd.DataFrame(instance_rows)
    manifest = pd.DataFrame(manifest_rows)
    fixed_orders = pd.DataFrame(order_rows)

    # 检查随机图数量和配对实例数量。
    expected_graphs = (
        len(config.node_sizes)
        * len(config.target_mean_indegrees)
        * config.graphs_per_cell
    )
    expected_instances = expected_graphs * config.n_fixed_orders

    if len(manifest) != expected_graphs:
        raise AssertionError(
            f"随机图数错误：得到{len(manifest)}，预期{expected_graphs}。"
        )
    if len(details) != expected_instances:
        raise AssertionError(
            f"一致性实例数错误：得到{len(details)}，"
            f"预期{expected_instances}。"
        )

    # 按(n,d)组合汇总一致率。
    summary_rows: list[dict[str, object]] = []
    group_columns = ["n_nodes", "target_mean_indegree"]

    for (n_nodes, target_d), group in details.groupby(
        group_columns,
        sort=False,
    ):
        # 所有实例进入分母，报错和缺失统一按False处理。
        def rate(column: str) -> float:
            values = group[column].fillna(False).astype(bool)
            return 100.0 * float(values.mean())

        n_successful = int((group["status"] == "ok").sum())

        # 保存单元格级汇总结果。
        summary_rows.append(
            {
                "n_nodes": int(n_nodes),
                "target_mean_indegree": float(target_d),
                "n_random_graphs": group["graph_id"].nunique(),
                "n_paired_instances": len(group),
                "mean_actual_edges": float(group["n_edges"].mean()),
                "n_successful": n_successful,
                "n_errors": len(group) - n_successful,
                "fill_equal_rate_pct": rate("fill_equal"),
                "graph_equal_rate_pct": rate("graph_equal"),
                "output_consistency_rate_pct": rate(
                    "output_consistency_passed"
                ),
                "structurally_valid_rate_pct": rate(
                    "structurally_valid"
                ),
                "n_output_mismatches": int(
                    (
                        ~group["output_consistency_passed"]
                        .fillna(False)
                    ).sum()
                ),
            }
        )

    # 返回汇总、实例、输入和完整失败证据。
    summary = pd.DataFrame(summary_rows)
    return (
        summary,
        details,
        manifest,
        fixed_orders,
        mismatch_evidence_rows,
    )


# ==========================================================
# 7. 实验 B：540 个独立运行时间实例
# ==========================================================





def _warm_up_cell(
    case: RandomGraphCase,
    repetitions: int,
) -> None:
    """每个(n,d)组合只执行算法预热，不做输入或输出验证。"""
    C_set = set(case.C)
    beta_tuple = tuple(case.orders[0].beta)
    for warmup_id in range(1, repetitions + 1):
        gc.collect()
        _run_pair_without_gc(
            G=case.G,
            C_set=C_set,
            beta=beta_tuple,
            fcgr_first=warmup_id % 2 == 1,
        )




def run_one_random_runtime_case(
    case: RandomGraphCase,
    timing_repetitions: int,
) -> tuple[dict[str, object], list[dict[str, object]]]:
    """
    对一张随机图的beta_1执行纯算法配对计时。

    不做基准验证或重复后输出比较；FCGR最终填充边集仅在全部计时
    结束后计算一次。
    """
    order = case.orders[0]
    C_set = set(case.C)
    beta_tuple = tuple(order.beta)
    algorithm1_times_ns: list[int] = []
    algorithm2_times_ns: list[int] = []
    fcgr_total_times_ns: list[int] = []
    ros_total_times_ns: list[int] = []
    repeat_rows: list[dict[str, object]] = []
    last_fcgr_result: FCGRResult | None = None
    last_ros_edges: list[Edge] | None = None

    for repeat_id in range(1, timing_repetitions + 1):
        gc.collect()
        fcgr_first = (case.graph_id + repeat_id) % 2 == 0
        (
            fcgr_result,
            ros_result,
            algorithm1_ns,
            algorithm2_ns,
            ros_total_ns,
        ) = _run_pair_without_gc(
            G=case.G,
            C_set=C_set,
            beta=beta_tuple,
            fcgr_first=fcgr_first,
        )
        fcgr_total_ns = algorithm1_ns + algorithm2_ns
        algorithm1_times_ns.append(algorithm1_ns)
        algorithm2_times_ns.append(algorithm2_ns)
        fcgr_total_times_ns.append(fcgr_total_ns)
        ros_total_times_ns.append(ros_total_ns)
        repeat_rows.append(
            {
                "n_nodes": case.n_nodes,
                "target_mean_indegree": case.target_mean_indegree,
                "graph_id": case.graph_id,
                "repeat_id": repeat_id,
                "timing_scope": "pure_algorithm_only",
                "run_order": (
                    "FCGR_then_ROS"
                    if fcgr_first
                    else "ROS_then_FCGR"
                ),
                "FCGR_algorithm1_ms": algorithm1_ns / 1_000_000,
                "FCGR_algorithm2_ms": algorithm2_ns / 1_000_000,
                "FCGR_total_ms": fcgr_total_ns / 1_000_000,
                "ROS_total_ms": ros_total_ns / 1_000_000,
                "runtime_ratio_FCGR_to_ROS": (
                    fcgr_total_ns / ros_total_ns
                ),
            }
        )
        last_fcgr_result = fcgr_result
        last_ros_edges = ros_result

    if last_fcgr_result is None or last_ros_edges is None:
        raise RuntimeError("该随机图没有产生任何正式计时。")

    algorithm1_median_ns = float(median(algorithm1_times_ns))
    algorithm2_median_ns = float(median(algorithm2_times_ns))
    fcgr_total_median_ns = float(median(fcgr_total_times_ns))
    ros_total_median_ns = float(median(ros_total_times_ns))

    # 明确位于全部计时区间之外。
    fcgr_fill_count = len(
        _fcgr_added_edges_R(
            G=case.G,
            C=C_set,
            H_R=last_fcgr_result.H_R,
        )
    )
    ros_fill_count = len(last_ros_edges)

    detail_row: dict[str, object] = {
        "n_nodes": case.n_nodes,
        "target_mean_indegree": case.target_mean_indegree,
        "graph_id": case.graph_id,
        "generation_attempt": case.generation_attempt,
        "graph_seed": case.graph_seed,
        "condition_seed": case.condition_seed,
        "beta_seed": order.beta_seed,
        "graph_fingerprint_sha256": _graph_fingerprint(case.G),
        "C_sha256": _hash_nodes(case.C),
        "beta_sha256": _hash_order(order.beta),
        "n_edges": case.G.number_of_edges(),
        "actual_mean_indegree": (
            case.G.number_of_edges() / case.G.number_of_nodes()
        ),
        "c_size": len(C_set),
        "timing_scope": "pure_algorithm_only",
        "timing_repetitions": timing_repetitions,
        "FCGR_algorithm1_median_ms": algorithm1_median_ns / 1_000_000,
        "FCGR_algorithm2_median_ms": algorithm2_median_ns / 1_000_000,
        "FCGR_total_median_ms": fcgr_total_median_ns / 1_000_000,
        "ROS_total_median_ms": ros_total_median_ns / 1_000_000,
        "runtime_ratio_FCGR_to_ROS": (
            fcgr_total_median_ns / ros_total_median_ns
        ),
        "FCGR_fill_count_after_timing": fcgr_fill_count,
        "ROS_fill_count": ros_fill_count,
    }
    return detail_row, repeat_rows



def _paired_bootstrap_runtime_ratio_ci(
    fcgr_times: np.ndarray,
    ros_times: np.ndarray,
    repetitions: int,
    seed: int,
) -> tuple[float, float]:
    """
    以随机图为单位计算配对bootstrap置信区间。

    每次重采样计算：
        mean(FCGR) / mean(ROS)。
    """
    # 检查两种算法的配对样本长度。
    if len(fcgr_times) != len(ros_times):
        raise ValueError("FCGR与ROS的配对样本长度不一致。")

    # 空样本返回缺失值。
    if len(fcgr_times) == 0:
        return float("nan"), float("nan")

    # 根据固定种子生成配对重采样索引。
    rng = np.random.default_rng(seed)
    sample_indices = rng.integers(
        0,
        len(fcgr_times),
        size=(repetitions, len(fcgr_times)),
    )

    # 在每次bootstrap样本中分别计算两种算法的平均时间。
    sampled_fcgr = fcgr_times[sample_indices].mean(axis=1)
    sampled_ros = ros_times[sample_indices].mean(axis=1)

    # 时间比统一定义为FCGR/ROS。
    ratios = sampled_fcgr / sampled_ros

    # 返回2.5%和97.5%分位数。
    lower, upper = np.quantile(ratios, [0.025, 0.975])
    return float(lower), float(upper)



def _summarize_random_runtime(
    details: pd.DataFrame,
    config: RandomRuntimeConfig,
) -> pd.DataFrame:
    """按(n,d)组合汇总30张随机图的纯算法代表时间。"""
    summary_rows: list[dict[str, object]] = []
    for (n_nodes, target_d), group in details.groupby(
        ["n_nodes", "target_mean_indegree"],
        sort=False,
    ):
        algorithm1_mean = float(
            group["FCGR_algorithm1_median_ms"].mean()
        )
        algorithm2_mean = float(
            group["FCGR_algorithm2_median_ms"].mean()
        )
        fcgr_total_mean = float(
            group["FCGR_total_median_ms"].mean()
        )
        ros_total_mean = float(
            group["ROS_total_median_ms"].mean()
        )
        bootstrap_seed = _derive_seed(
            config.root_seed,
            "bootstrap",
            n_nodes,
            target_d,
        )
        ci_lower, ci_upper = _paired_bootstrap_runtime_ratio_ci(
            fcgr_times=group["FCGR_total_median_ms"].to_numpy(
                dtype=float
            ),
            ros_times=group["ROS_total_median_ms"].to_numpy(
                dtype=float
            ),
            repetitions=config.bootstrap_repetitions,
            seed=bootstrap_seed,
        )
        summary_rows.append(
            {
                "n_nodes": int(n_nodes),
                "target_mean_indegree": float(target_d),
                "n_random_graphs": len(group),
                "mean_actual_edges": float(group["n_edges"].mean()),
                "mean_actual_indegree": float(
                    group["actual_mean_indegree"].mean()
                ),
                "timing_scope": "pure_algorithm_only",
                "FCGR_algorithm1_mean_ms": algorithm1_mean,
                "FCGR_algorithm2_mean_ms": algorithm2_mean,
                "FCGR_total_mean_ms": fcgr_total_mean,
                "ROS_total_mean_ms": ros_total_mean,
                "runtime_ratio_FCGR_to_ROS": (
                    fcgr_total_mean / ros_total_mean
                ),
                "runtime_ratio_CI95_lower": ci_lower,
                "runtime_ratio_CI95_upper": ci_upper,
            }
        )
    return pd.DataFrame(summary_rows)




def run_random_runtime_experiment(
    config: RandomRuntimeConfig = RandomRuntimeConfig(),
) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """独立运行随机网络纯算法计时实验。"""
    _validate_design_config(config)
    detail_rows: list[dict[str, object]] = []
    repeat_rows: list[dict[str, object]] = []
    manifest_rows: list[dict[str, object]] = []

    for n_nodes in config.node_sizes:
        for target_d in config.target_mean_indegrees:
            cases = [
                build_random_graph_case(
                    n_nodes=n_nodes,
                    target_mean_indegree=target_d,
                    graph_id=graph_id,
                    config=config,
                )
                for graph_id in range(1, config.graphs_per_cell + 1)
            ]
            _warm_up_cell(
                case=cases[0],
                repetitions=config.warmup_repetitions,
            )

            for case in cases:
                manifest_rows.append(_case_manifest_row(case))
                detail_row, current_repeat_rows = (
                    run_one_random_runtime_case(
                        case=case,
                        timing_repetitions=config.timing_repetitions,
                    )
                )
                detail_rows.append(detail_row)
                repeat_rows.extend(current_repeat_rows)

            current_details = pd.DataFrame(
                row
                for row in detail_rows
                if (
                    row["n_nodes"] == n_nodes
                    and row["target_mean_indegree"] == target_d
                )
            )
            current_summary = _summarize_random_runtime(
                current_details,
                config,
            ).iloc[0]
            print(
                "[计时] "
                f"n={n_nodes}, d={target_d:g}："
                f"FCGR={current_summary['FCGR_total_mean_ms']:.6f} ms，"
                f"ROS={current_summary['ROS_total_mean_ms']:.6f} ms。"
            )

    details = pd.DataFrame(detail_rows)
    raw_times = pd.DataFrame(repeat_rows)
    manifest = pd.DataFrame(manifest_rows)
    summary = _summarize_random_runtime(details, config)
    expected_instances = (
        len(config.node_sizes)
        * len(config.target_mean_indegrees)
        * config.graphs_per_cell
    )
    expected_raw_times = expected_instances * config.timing_repetitions
    if len(details) != expected_instances:
        raise AssertionError(
            f"计时实例数错误：得到{len(details)}，"
            f"预期{expected_instances}。"
        )
    if len(raw_times) != expected_raw_times:
        raise AssertionError(
            f"原始计时行数错误：得到{len(raw_times)}，"
            f"预期{expected_raw_times}。"
        )
    return summary, details, raw_times, manifest



# ==========================================================
# 8. 结果保存与运行时间图
# ==========================================================


def _config_frame(
    config: RandomConsistencyConfig | RandomRuntimeConfig,
) -> pd.DataFrame:
    rows: list[dict[str, object]] = []
    for parameter, value in asdict(config).items():
        if isinstance(value, (tuple, list, dict)):
            value = json.dumps(value, ensure_ascii=False)
        rows.append({"parameter": parameter, "value": value})
    return pd.DataFrame(rows)


def _environment_frame() -> pd.DataFrame:
    versions = {
        "timestamp_local": time.strftime("%Y-%m-%d %H:%M:%S"),
        "platform": platform.platform(),
        "python": sys.version.replace("\n", " "),
        "networkx": nx.__version__,
        "numpy": np.__version__,
        "pandas": pd.__version__,
    }
    return pd.DataFrame(
        {
            "item": list(versions.keys()),
            "value": list(versions.values()),
        }
    )


def save_random_consistency_results(
    summary: pd.DataFrame,
    details: pd.DataFrame,
    manifest: pd.DataFrame,
    fixed_orders: pd.DataFrame,
    mismatch_evidence: Sequence[Mapping[str, object]],
    config: RandomConsistencyConfig,
    output_file: str | Path,
    mismatch_file: str | Path,
) -> tuple[Path, Path]:
    """
    保存一致性结果。

    Excel只保存边数与SHA-256，避免大网络边集超过单元格上限；
    完整不一致证据保存到 gzip 压缩 JSONL。
    """
    output_file = Path(output_file)
    mismatch_file = Path(mismatch_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)
    mismatch_file.parent.mkdir(parents=True, exist_ok=True)

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(writer, sheet_name="cell_summary", index=False)
        details.to_excel(
            writer,
            sheet_name="instance_results",
            index=False,
        )
        manifest.to_excel(
            writer,
            sheet_name="input_manifest",
            index=False,
        )
        fixed_orders.to_excel(
            writer,
            sheet_name="fixed_orders",
            index=False,
        )
        _config_frame(config).to_excel(
            writer,
            sheet_name="config",
            index=False,
        )
        _environment_frame().to_excel(
            writer,
            sheet_name="environment",
            index=False,
        )

    with gzip.open(mismatch_file, mode="wt", encoding="utf-8") as handle:
        for evidence in mismatch_evidence:
            handle.write(
                json.dumps(
                    evidence,
                    ensure_ascii=False,
                    separators=(",", ":"),
                )
            )
            handle.write("\n")

    return output_file, mismatch_file


def save_random_runtime_results(
    summary: pd.DataFrame,
    details: pd.DataFrame,
    raw_times: pd.DataFrame,
    manifest: pd.DataFrame,
    config: RandomRuntimeConfig,
    output_file: str | Path,
) -> Path:
    output_file = Path(output_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        summary.to_excel(writer, sheet_name="cell_summary", index=False)
        details.to_excel(
            writer,
            sheet_name="graph_details",
            index=False,
        )
        raw_times.to_excel(
            writer,
            sheet_name="raw_repeat_times",
            index=False,
        )
        manifest.to_excel(
            writer,
            sheet_name="input_manifest",
            index=False,
        )
        _config_frame(config).to_excel(
            writer,
            sheet_name="config",
            index=False,
        )
        _environment_frame().to_excel(
            writer,
            sheet_name="environment",
            index=False,
        )

    return output_file


def save_runtime_figure(
    summary: pd.DataFrame,
    output_file: str | Path,
) -> Path:
    """
    按目标平均入度分为三个子图，横纵轴均使用对数刻度。
    """
    output_file = Path(output_file)
    output_file.parent.mkdir(parents=True, exist_ok=True)
    target_ds = list(summary["target_mean_indegree"].drop_duplicates())
    fig, axes = plt.subplots(
        1,
        len(target_ds),
        figsize=(5.2 * len(target_ds), 4.6),
        sharey=True,
        constrained_layout=True,
    )
    if len(target_ds) == 1:
        axes = [axes]

    for axis, target_d in zip(axes, target_ds):
        group = (
            summary[
                summary["target_mean_indegree"] == target_d
            ]
            .sort_values("n_nodes")
            .copy()
        )
        axis.plot(
            group["n_nodes"],
            group["FCGR_total_mean_ms"],
            marker="o",
            linewidth=1.8,
            label="FCGR",
        )
        axis.plot(
            group["n_nodes"],
            group["ROS_total_mean_ms"],
            marker="s",
            linewidth=1.8,
            label="ROS",
        )
        axis.set_xscale("log")
        axis.set_yscale("log")
        axis.set_title(
            f"Target mean indegree = {target_d:g}"
        )
        axis.set_xlabel("Number of nodes n")
        axis.grid(True, which="both", linestyle="--", alpha=0.35)

    axes[0].set_ylabel("Mean runtime (ms, log scale)")
    axes[-1].legend(frameon=False)
    fig.savefig(output_file, dpi=300, bbox_inches="tight")
    plt.close(fig)
    return output_file


```

```python
# 随机网络实验配置：2160个一致性实例与540个计时实例彼此独立。
RANDOM_CONSISTENCY_CONFIG = RandomConsistencyConfig(
    node_sizes=(20, 50, 100, 200, 500, 1000),
    target_mean_indegrees=(1.0, 2.0, 4.0),
    graphs_per_cell=30,
    c_size=3,
    n_fixed_orders=4,
    root_seed=20260730,
    topological_order_attempts=10_000,
    graph_retry_limit=100,
    stop_on_mismatch=False,
)

RANDOM_RUNTIME_CONFIG = RandomRuntimeConfig(
    node_sizes=(20, 50, 100, 200, 500, 1000),
    target_mean_indegrees=(1.0, 2.0, 4.0),
    graphs_per_cell=30,
    c_size=3,
    n_fixed_orders=4,
    root_seed=20260730,
    topological_order_attempts=10_000,
    graph_retry_limit=100,
    warmup_repetitions=5,
    timing_repetitions=30,
    bootstrap_repetitions=10_000,
)

if _design_signature(RANDOM_CONSISTENCY_CONFIG) != _design_signature(
    RANDOM_RUNTIME_CONFIG
):
    raise ValueError(
        "随机网络一致性实验与计时实验的输入设计不一致；"
        "无法保证由同一根种子重建同一批540张图。"
    )

RANDOM_RESULT_DIR = Path("./results_random_networks")
RANDOM_CONSISTENCY_RESULT_FILE = (
    RANDOM_RESULT_DIR / "FCGR_ROS_random_consistency_2160_cases.xlsx"
)
RANDOM_MISMATCH_EVIDENCE_FILE = (
    RANDOM_RESULT_DIR / "FCGR_ROS_random_mismatches.jsonl.gz"
)
RANDOM_RUNTIME_RESULT_FILE = (
    RANDOM_RESULT_DIR / "FCGR_ROS_random_runtime_540_cases.xlsx"
)
RANDOM_RUNTIME_FIGURE_FILE = (
    RANDOM_RESULT_DIR / "FCGR_ROS_random_runtime_loglog.png"
)


def run_random_consistency_only(
    config: RandomConsistencyConfig = RANDOM_CONSISTENCY_CONFIG,
    output_file: str | Path = RANDOM_CONSISTENCY_RESULT_FILE,
    mismatch_file: str | Path = RANDOM_MISMATCH_EVIDENCE_FILE,
) -> dict[str, object]:
    """
    独立运行随机网络输出一致性实验。

    主通过标准仅为填充边集一致与最终输出图一致。
    """
    # 运行全部随机图与四个固定拓扑序的配对实验。
    (
        summary,
        details,
        manifest,
        fixed_orders,
        mismatch_evidence,
    ) = run_random_consistency_experiment(config=config)

    # 保存Excel结果和完整不一致证据。
    saved_path, mismatch_path = save_random_consistency_results(
        summary=summary,
        details=details,
        manifest=manifest,
        fixed_orders=fixed_orders,
        mismatch_evidence=mismatch_evidence,
        config=config,
        output_file=output_file,
        mismatch_file=mismatch_file,
    )

    # 显示核心结果和文件位置。
    print(summary.to_string(index=False))
    print(f"输出不一致或报错实例数：{len(mismatch_evidence)}")
    print(f"结果文件：{saved_path.resolve()}")
    print(f"完整不一致证据：{mismatch_path.resolve()}")

    # 返回全部实验结果。
    return {
        "summary": summary,
        "details": details,
        "manifest": manifest,
        "fixed_orders": fixed_orders,
        "mismatch_evidence": mismatch_evidence,
        "output_file": saved_path,
        "mismatch_file": mismatch_path,
    }


def run_random_runtime_only(
    config: RandomRuntimeConfig = RANDOM_RUNTIME_CONFIG,
    output_file: str | Path = RANDOM_RUNTIME_RESULT_FILE,
    figure_file: str | Path = RANDOM_RUNTIME_FIGURE_FILE,
) -> dict[str, object]:
    """
    独立运行随机网络无验证计时实验。

    输入验证、结果审计与输出比较均在计时区间之外完成。
    """
    # 运行540张随机图的无验证配对计时。
    summary, details, raw_times, manifest = (
        run_random_runtime_experiment(config=config)
    )

    # 保存汇总、随机图级结果、原始重复时间和输入清单。
    saved_path = save_random_runtime_results(
        summary=summary,
        details=details,
        raw_times=raw_times,
        manifest=manifest,
        config=config,
        output_file=output_file,
    )

    # 保存FCGR与ROS运行时间随规模变化的对数图。
    figure_path = save_runtime_figure(
        summary=summary,
        output_file=figure_file,
    )

    # 显示主结果和输出位置。
    print(summary.to_string(index=False))
    print(f"结果文件：{saved_path.resolve()}")
    print(f"运行时间图：{figure_path.resolve()}")

    # 返回全部结果对象。
    return {
        "summary": summary,
        "details": details,
        "raw_times": raw_times,
        "manifest": manifest,
        "output_file": saved_path,
        "figure_file": figure_path,
    }
```
