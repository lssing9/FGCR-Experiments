```python
import networkx as nx
from itertools import combinations
from typing import Hashable, Iterable, Sequence
Node = Hashable
Edge = tuple[Node, Node]
```

```python
# 输入验证
def _validate_dag(G: nx.DiGraph) -> None:
    """
    检查 G 是否为简单有向无环图。
    """
    if not isinstance(G, nx.DiGraph):
        raise TypeError("G 必须是 networkx.DiGraph。")
    if G.is_multigraph():
        raise TypeError("G 必须是简单有向图，不能是 MultiDiGraph。")
    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("输入图 G 必须是 DAG。")

def _validate_conditioning_set(
    G: nx.DiGraph,
    C: Iterable[Node],
) -> set[Node]:
    """
    检查条件集 C 是否为 V 的非空真子集。
    """
    C_set = set(C)
    V_set = set(G.nodes)

    if not C_set:
        raise ValueError("条件集 C 不能为空。")
    if not C_set < V_set:
        raise ValueError("条件集 C 必须是 V 的非空真子集。")
    return C_set

def _validate_topological_order(
    G: nx.DiGraph,
    alpha: Sequence[Node],
) -> tuple[list[Node], dict[Node, int]]:
    """
    检查 alpha 是否恰好包含 G 的全部节点一次， 并且是 G 的合法拓扑序。
    返回：
        alpha_list：
            alpha 的列表形式。
        alpha_position：
            节点到其顺序位置的映射。
    """
    alpha_list = list(alpha)
    V_set = set(G.nodes)
    if len(alpha_list) != G.number_of_nodes() or set(alpha_list) != V_set:
        raise ValueError("alpha 必须恰好包含 G 中的每个节点一次。")
    alpha_position = {node: index for index, node in enumerate(alpha_list)}
    for source, target in G.edges:
        if alpha_position[source] >= alpha_position[target]:
            raise ValueError("alpha 不是原图 G 的合法拓扑序：" f"边 {source!r}->{target!r} 与该顺序冲突。")
    return alpha_list, alpha_position

#返回 node 的全部祖先，并包含 node 自身。
def ancestors_with_self(
    G: nx.DiGraph,
    node: Node,
) -> set[Node]:
    if node not in G:
        raise ValueError(  f"节点 {node!r} 不在图 G 中。"   )
    ancestors = set(nx.ancestors(G, node))
    ancestors.add(node)
    return ancestors
```

```python
# ROS 算法

def minimal_fill_edge_set(
    G: nx.DiGraph,
    alpha: Sequence[Node],
    C: Iterable[Node],
) -> dict:
    """
    ROS 算法：寻找条件化模型的最小填充边集。
        E_alpha_R:
            ROS 找到的新增有向边，每条边保存为 (source, target)，
        G_alpha_R:
            G_R 加上 E_alpha_R 后得到的条件模型 minimal I-map。
    """
   
    # 0. 输入验证
    _validate_dag(G)
    C_set = _validate_conditioning_set(G=G, C=C)
    alpha_list, alpha_position = _validate_topological_order(G=G, alpha=alpha)
    # 保存原始图，不修改输入对象
    G_original = G
    G_work = G.copy()
    
    # 1. 构造 G^C：删除 C 中节点的全部出边
    for conditioned_node in C_set:
        outgoing_edges = [(conditioned_node, child) for child in G_work.successors(conditioned_node)]
        G_work.remove_edges_from(outgoing_edges)
    
    # 2. 计算 An_{G^C}(C)
    nodes_of_interest: set[Node] = set()
    for conditioned_node in C_set:
        nodes_of_interest.update(ancestors_with_self(G=G_work, node=conditioned_node))
    
    # 3. 构造祖先闭包上的诱导子图 G_0
    G_0 = G_work.subgraph(nodes_of_interest).copy()
    G_0.name = f"{G.name or 'G'}_ROS_search"
    
    # 4. 按 alpha 的逆序排列搜索节点
    vertices = [node for node in reversed(alpha_list) if node in nodes_of_interest]
    k = len(vertices)

    # 每条新增边保存为真正的节点元组
    E_alpha_R: list[Edge] = []
 
    # 5. 逐层完成父节点集合
    if k >= 3:
        for current_node in vertices[: k - 2]:
            parents = sorted(G_0.predecessors(current_node), key=alpha_position.__getitem__)
            if len(parents) < 2:
                continue
            for x, y in combinations(parents, 2):
                # 已经邻接时不重复添加
                if G_0.has_edge(x, y) or G_0.has_edge(y, x):
                    continue
                # 按 alpha 从左向右定向
                new_edge = (x, y) if alpha_position[x] < alpha_position[y] else (y, x)
                G_0.add_edge(*new_edge)
                E_alpha_R.append(new_edge)

    # 6. 构造原图在 R=V\\C 上的完整导出子图
    R_nodes = [node for node in G_original.nodes if node not in C_set]

    G_alpha_R = G_original.subgraph(R_nodes).copy()
    G_alpha_R.name = f"{G.name or 'G'}_ROS_HR"
    
    # 7. 添加 ROS 找到的填充边
    G_alpha_R.add_edges_from(E_alpha_R)
    
    # 8. 输出检查
    if not nx.is_directed_acyclic_graph(G_alpha_R):
        raise AssertionError("ROS 输出图 G_alpha_R 不是 DAG。")

    for source, target in G_alpha_R.edges:
        if alpha_position[source] >= alpha_position[target]:
            raise AssertionError("ROS 输出边与 alpha 不一致：" f"{source!r}->{target!r}。")

    return {"E_alpha_R": E_alpha_R, "G_alpha_R": G_alpha_R}
```

```python
G = nx.DiGraph()
G.add_edges_from([ ("a", "c"), ("b", "d"), ("c", "e"), ("d", "e"),])
alpha = [ "a", "b", "c", "d", "e",]
C = {"e"}

ros_result = minimal_fill_edge_set( G=G,  alpha=alpha, C=C, )
print("ROS 新增边：")
print(ros_result["E_alpha_R"])

print("ROS 输出图的边：")
print(sorted(ros_result["G_alpha_R"].edges()))
```

```python
# 构造一个简单 DAG（示例）
G = nx.DiGraph()
G.add_edges_from([('a', 'c'), ('b', 'c')])
alpha = ['a', 'b', 'c']   # 一个合法的拓扑序
C = {'c'}                      # 条件节点

result = minimal_fill_edge_set(G, alpha, C)
print("Added edges:", result['E_alpha_R'])
print("Nodes in final graph:", list(result['G_alpha_R'].nodes))
```
